# Cálculo de Husos Horarios (Hz, Hc, Ho)
Simulador de cálculos de horas para navegación astronómica.

Este simulador está diseñado para fines educativos. **No lo utilices para la navegación real.**

```
import ipywidgets as widgets
from IPython.display import display
from datetime import datetime, timedelta

def calcular_horas(hcg_str, longitud, zona_horaria_oficial):
    try:
        hcg = datetime.strptime(hcg_str, "%H:%M")
        
        # Longitud en grados decimales (positivo Este, negativo Oeste)
        hcl_offset = longitud / 15.0
        hcl = hcg + timedelta(hours=hcl_offset)
        
        # Huso (Hz o Z). Z = - (Longitud / 15) redondeado al entero más próximo
        hz_offset = round(longitud / 15.0)
        hz_time = hcg + timedelta(hours=hz_offset)
        
        # Hora Oficial (Ho)
        ho = hcg + timedelta(hours=zona_horaria_oficial)
        
        print(f"Hora Civil de Greenwich (HcG): {hcg.strftime('%H:%M:%S')}")
        print(f"Hora Civil de Lugar (HcL):   {hcl.strftime('%H:%M:%S')} (offset exacto de {hcl_offset:.2f}h)")
        print(f"Hora Legal (Hz):           {hz_time.strftime('%H:%M:%S')} (Huso Z: {hz_offset})")
        print(f"Hora Oficial (Ho):         {ho.strftime('%H:%M:%S')} (GMT+{zona_horaria_oficial})")
        
    except Exception as e:
        print("Error en los datos de entrada.")

hcg_in = widgets.Text(value="14:00", description='HcG (HH:MM):')
lon_in = widgets.FloatSlider(value=-10.0, min=-180.0, max=180.0, step=0.1, description='Longitud (º):')
zo_in = widgets.IntSlider(value=2, min=-12, max=12, step=1, description='Huso Oficial:')

out = widgets.interactive_output(calcular_horas, {'hcg_str': hcg_in, 'longitud': lon_in, 'zona_horaria_oficial': zo_in})
display(widgets.VBox([hcg_in, lon_in, zo_in, out]))
```
