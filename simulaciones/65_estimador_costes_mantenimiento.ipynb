# Estimador de Costes Anuales del Buque
Cálculo orientativo del TCO (Total Cost of Ownership) para mantener una embarcación.

Este simulador está diseñado para fines educativos. **No lo utilices para la navegación real.**

```
import ipywidgets as widgets
from IPython.display import display

def estimar_costes(eslora, es_velero, edad_barco, uso_intensivo):
    # Estimación grosera basada en tarifas medias (España)
    
    coste_amarre_anual = eslora * (eslora/3) * 120 # F(eslora, manga aprox)
    
    coste_seguro = 200 + (eslora * 40)
    if eslora > 12:
        coste_seguro *= 1.5
        
    # Varadero anual: limpieza, patente (antifouling), grúa
    coste_varadero = 300 + (eslora * 80)
    
    # Revisión motor: Aceite, filtros, impeller
    coste_motor = 400 if es_velero else 800
    if uso_intensivo:
        coste_motor *= 1.5
        
    # Imprevistos y renovación (Jarcia, velas, bombas, electrónica) 
    # Mayor cuanto más viejo y más grande
    factor_edad = min(edad_barco / 10, 2)
    coste_imprevistos = eslora * 150 * (1 + factor_edad)
    
    total = coste_amarre_anual + coste_seguro + coste_varadero + coste_motor + coste_imprevistos
    
    print(f"--- Estimador de Costes Anuales (Eslora: {eslora}m) ---")
    print(f"🚢 Amarre anual (estimado): {coste_amarre_anual:.0f} €")
    print(f"🛡️ Seguro a terceros / Todo riesgo básico: {coste_seguro:.0f} €")
    print(f"🏗️ Varadero y Antifouling (anual): {coste_varadero:.0f} €")
    print(f"⚙️ Mantenimiento Motor básico: {coste_motor:.0f} €")
    print(f"🔧 Reserva Imprevistos / Desgaste: {coste_imprevistos:.0f} €")
    print("-------------------------------------------------------")
    print(f"💰 COSTE TOTAL ANUAL APROXIMADO: {total:.0f} €")
    print(f"   (Coste medio mensual: {total/12:.0f} €/mes)")
    print("\n*Nota: Estos valores son estadísticos y educativos, pueden variar inmensamente según el puerto y el bricolaje del armador (DIY).*")

es = widgets.FloatSlider(value=10.0, min=5.0, max=24.0, step=0.5, description='Eslora (m):')
vela = widgets.Checkbox(value=True, description='Es Velero (vs Motora)')
edad = widgets.IntSlider(value=10, min=0, max=50, step=1, description='Edad (años):')
uso = widgets.Checkbox(value=False, description='Uso Intensivo / Chárter')

out = widgets.interactive_output(estimar_costes, {'eslora': es, 'es_velero': vela, 'edad_barco': edad, 'uso_intensivo': uso})
display(widgets.VBox([es, vela, edad, uso, out]))
```
