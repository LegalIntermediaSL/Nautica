# Modelo de Supervivencia: Tiempo de Hipotermia

Al caer al mar, el cuerpo pierde calor 25 veces más rápido que en el aire. Calcularemos el *Tiempo de Pérdida de Conciencia* (TUC) y el *Tiempo de Supervivencia*.


```python
import matplotlib.pyplot as plt

def curva_hipotermia(temp_agua_c):
    # Datos aproximados de la tabla estándar de USCG (United States Coast Guard)
    temps = [2, 5, 10, 15, 20]
    consciencia = [0.25, 0.5, 1.0, 4.0, 10.0] # Horas
    supervivencia = [0.75, 1.5, 3.0, 10.0, 24.0] # Horas
    
    plt.figure(figsize=(7,4))
    plt.plot(temps, consciencia, 'y-o', label='Tiempo de Conciencia')
    plt.plot(temps, supervivencia, 'r-s', label='Tiempo de Supervivencia Máxima')
    plt.axvline(x=temp_agua_c, color='b', linestyle='--', label=f'Temp. Actual {temp_agua_c}ºC')
    plt.title("Hipotermia Inmersión (Sin traje térmico)")
    plt.xlabel("Temperatura del Agua (ºC)")
    plt.ylabel("Horas")
    plt.grid(True)
    plt.legend()
    plt.show()

curva_hipotermia(12) # Atlántico Norte/Cantábrico
```
