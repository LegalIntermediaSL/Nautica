# Estabilidad Dinámica y Resonancia
Efecto de la frecuencia de las olas sobre el periodo de balanceo del buque.

```
import ipywidgets as widgets
from IPython.display import display
import numpy as np
import matplotlib.pyplot as plt

def simulador_resonancia(manga, gm, periodo_olas):
    # Periodo propio de balanceo aproximado (segundos): T = 0.8 * Manga / sqrt(GM)
    if gm <= 0:
        print("El barco es inestable (GM <= 0). Volcará de inmediato.")
        return
        
    periodo_barco = 0.8 * manga / np.sqrt(gm)
    
    print(f"Periodo de Balanceo Natural del Barco: {periodo_barco:.1f} segundos")
    print(f"Periodo del Oleaje (Encuentro): {periodo_olas:.1f} segundos")
    
    # Simularemos 30 segundos
    t = np.linspace(0, 30, 300)
    
    # Amortiguamiento
    damp = 0.05
    
    # Respuesta (Aproximación de oscilador armónico forzado)
    w_b = 2 * np.pi / periodo_barco
    w_o = 2 * np.pi / periodo_olas
    
    amplitud_ola = 10 # Grados de pendiente de la ola
    
    # Magnificación
    den = np.sqrt((w_b**2 - w_o**2)**2 + (2*damp*w_b*w_o)**2)
    amp_barco = (w_b**2 * amplitud_ola) / den
    
    # Respuesta en el tiempo (Simplificada estado estacionario)
    balanceo = amp_barco * np.sin(w_o * t)
    
    fig, ax = plt.subplots(figsize=(8,4))
    ax.plot(t, balanceo, 'r-', label='Ángulo de Escora del Barco')
    ax.plot(t, amplitud_ola * np.sin(w_o * t), 'b--', alpha=0.5, label='Pendiente de la Ola')
    
    ax.set_ylabel('Grados')
    ax.set_xlabel('Tiempo (s)')
    ax.legend()
    plt.grid(True)
    
    if abs(periodo_barco - periodo_olas) < 1.0:
        plt.title('⚠️ PELIGRO: RESONANCIA (Periodos Coinciden)')
        print("\n⚠️ RIESGO EXTREMO: El periodo de la ola coincide con el del barco. La amplitud de balanceo se multiplica pudiendo causar vuelco.")
    else:
        plt.title('Navegación Estable (Fuera de Resonancia)')
        
    plt.show()

m = widgets.FloatSlider(value=4.0, min=2, max=10, description='Manga (m):')
gm = widgets.FloatSlider(value=1.5, min=0.1, max=5, description='GM (m):')
po = widgets.FloatSlider(value=5.0, min=2, max=15, description='Periodo Olas:')

out = widgets.interactive_output(simulador_resonancia, {'manga': m, 'gm': gm, 'periodo_olas': po})
display(widgets.VBox([m, gm, po, out]))
```
