```
import ipywidgets as widgets
from IPython.display import display
import math
import matplotlib.pyplot as plt
import numpy as np

def resolver_cinematica(rumbo_propio, vel_propia, d1, dist1, d2, dist2, tiempo_minutos):
    # Pasar a radianes
    rp_rad = math.radians(90 - rumbo_propio)
    d1_rad = math.radians(90 - d1)
    d2_rad = math.radians(90 - d2)
    
    # Coordenadas relativas
    x1 = dist1 * math.cos(d1_rad)
    y1 = dist1 * math.sin(d1_rad)
    x2 = dist2 * math.cos(d2_rad)
    y2 = dist2 * math.sin(d2_rad)
    
    # Movimiento Relativo
    dx = x2 - x1
    dy = y2 - y1
    dist_rel = math.hypot(dx, dy)
    vel_rel = dist_rel / (tiempo_minutos / 60)
    
    rumbo_rel_rad = math.atan2(dy, dx)
    rumbo_rel = (90 - math.degrees(rumbo_rel_rad)) % 360
    
    # Vector propio (invertido para el triángulo de velocidades)
    e_x = -vel_propia * math.cos(rp_rad)
    e_y = -vel_propia * math.sin(rp_rad)
    
    # Vector relativo
    r_x = vel_rel * math.cos(rumbo_rel_rad)
    r_y = vel_rel * math.sin(rumbo_rel_rad)
    
    # Vector verdadero del otro buque
    v_x = r_x - e_x
    v_y = r_y - e_y
    vel_verdadera = math.hypot(v_x, v_y)
    rumbo_verdadero = (90 - math.degrees(math.atan2(v_y, v_x))) % 360
    
    # CPA (Punto de Máxima Aproximación)
    if vel_rel > 0:
        cpa_dist = abs(x1 * dy - y1 * dx) / dist_rel
    else:
        cpa_dist = 0
        
    print(f"Rumbo Verdadero del otro buque: {rumbo_verdadero:.1f}º")
    print(f"Velocidad Verdadera del otro buque: {vel_verdadera:.1f} nudos")
    print(f"Rumbo Relativo: {rumbo_rel:.1f}º a {vel_rel:.1f} nudos")
    print(f"CPA (Punto de máxima aproximación): {cpa_dist:.2f} millas")
    
    fig, ax = plt.subplots(figsize=(6,6), subplot_kw={'projection': 'polar'})
    ax.set_theta_zero_location('N')
    ax.set_theta_direction(-1)
    ax.plot(d1_rad, dist1, 'bo', label='E1')
    ax.plot(d2_rad, dist2, 'ro', label='E2')
    ax.plot([d1_rad, d2_rad], [dist1, dist2], 'r--')
    ax.legend()
    plt.show()

rp = widgets.FloatSlider(value=0, min=0, max=360, step=1, description='Rumbo Propio:')
vp = widgets.FloatSlider(value=10, min=0, max=30, step=1, description='Vel. Propia:')
d1 = widgets.FloatSlider(value=45, min=0, max=360, step=1, description='Demora 1:')
dst1 = widgets.FloatSlider(value=10, min=0, max=24, step=0.5, description='Distancia 1:')
d2 = widgets.FloatSlider(value=45, min=0, max=360, step=1, description='Demora 2:')
dst2 = widgets.FloatSlider(value=5, min=0, max=24, step=0.5, description='Distancia 2:')
tm = widgets.FloatSlider(value=15, min=1, max=60, step=1, description='Tiempo (m):')

ui = widgets.VBox([rp, vp, widgets.HBox([d1, dst1]), widgets.HBox([d2, dst2]), tm])
out = widgets.interactive_output(resolver_cinematica, {'rumbo_propio': rp, 'vel_propia': vp, 'd1': d1, 'dist1': dst1, 'd2': d2, 'dist2': dst2, 'tiempo_minutos': tm})
display(ui, out)
```
