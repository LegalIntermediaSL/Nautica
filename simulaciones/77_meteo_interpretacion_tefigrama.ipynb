# Termodinámica: Inestabilidad Atmosférica
Predicción de tormentas severas evaluando gradientes térmicos verticales.

```
import ipywidgets as widgets
from IPython.display import display

def interpretar_estabilidad(t_superficie, t_850hpa, t_500hpa):
    # Capas aproximadas: Superficie (0m), 850hPa (~1500m), 500hPa (~5500m)
    
    gradiente_inferior = (t_superficie - t_850hpa) / 1.5 # ºC por cada 1000m
    gradiente_superior = (t_850hpa - t_500hpa) / 4.0   # ºC por cada 1000m
    
    print("--- Análisis Termodinámico Simplificado (Perfil Vertical) ---")
    print(f"Gradiente térmico capa inferior (0-1500m): -{gradiente_inferior:.1f} ºC / km")
    print(f"Gradiente térmico capa media (1500-5500m): -{gradiente_superior:.1f} ºC / km\n")
    
    # DALR (Dry Adiabatic Lapse Rate) ~ 9.8 ºC/km
    # SALR (Saturated Adiabatic Lapse Rate) ~ 5-6 ºC/km
    
    # Análisis Capa Inferior
    if gradiente_inferior > 9.8:
        print("🔴 Capa Inferior: Inestabilidad ABSOLUTA. El aire asciende violentamente (Térmicas muy fuertes).")
    elif gradiente_inferior > 6.0:
        print("🟡 Capa Inferior: Inestabilidad CONDICIONAL. Si el aire está húmedo, se formarán cúmulos.")
    else:
        print("🟢 Capa Inferior: ESTABLE. No hay desarrollo vertical. Posible niebla o estratos.")
        
    # Análisis Tormentas Severas (Índice muy simplificado, "Lapse Rate")
    diferencia_total = t_superficie - t_500hpa
    print(f"\nDiferencia T(Sup) - T(500hPa): {diferencia_total:.1f} ºC")
    
    if diferencia_total > 35: # Si en altura hace mucho frío y abajo mucho calor
        print("🌩️ ALERTA DE TORMENTA SEVERA: El gradiente térmico es extremo. Si hay humedad, se desarrollarán Cumulonimbus gigantes con riesgo de turbonadas y granizo.")
    elif diferencia_total > 26:
        print("🌧️ Riesgo moderado de chubascos tormentosos.")
    else:
        print("☀️ Atmósfera globalmente estable. Buen tiempo para navegar.")

t_sup = widgets.FloatSlider(value=25.0, min=0, max=45, description='T Sup (ºC):')
t_850 = widgets.FloatSlider(value=10.0, min=-15, max=30, description='T 850hPa (ºC):')
t_500 = widgets.FloatSlider(value=-15.0, min=-40, max=10, description='T 500hPa (ºC):')

out = widgets.interactive_output(interpretar_estabilidad, {'t_superficie': t_sup, 't_850hpa': t_850, 't_500hpa': t_500})
display(widgets.VBox([t_sup, t_850, t_500, out]))
```
