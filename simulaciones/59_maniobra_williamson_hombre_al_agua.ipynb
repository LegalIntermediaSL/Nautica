```
import ipywidgets as widgets
from IPython.display import display
import matplotlib.pyplot as plt
import numpy as np

def williamson_turn(rumbo_inicial):
    print("Maniobra de Williamson (Hombre al Agua)")
    print(f"1. Timón a la banda de caída del náufrago.")
    caida1 = (rumbo_inicial + 60) % 360
    caida2 = (rumbo_inicial - 60) % 360
    print(f"2. Caer 60º del rumbo inicial. (Ejemplo Estribor: Rumbo a {caida1:.0f}º)")
    
    rumbo_reciproco = (rumbo_inicial + 180) % 360
    print(f"3. Meter todo el timón a la banda contraria.")
    print(f"4. Mantener la caída hasta llegar al rumbo recíproco: {rumbo_reciproco:.0f}º")
    print("5. Parar máquinas al aproximarse al náufrago por la amura de sotavento.")
    
    # Dibujo simplificado
    t = np.linspace(0, 1.5*np.pi, 100)
    x = np.sin(t) * t
    y = np.cos(t) * t
    
    plt.figure(figsize=(5,5))
    plt.plot(x, y, 'b-', label='Trayectoria del buque')
    plt.plot(0, 0, 'ro', label='Caída Hombre al agua')
    plt.axis('equal')
    plt.legend()
    plt.title('Trayectoria Curva de Williamson (Ilustrativo)')
    plt.grid(True)
    plt.show()

ri = widgets.IntSlider(value=45, min=0, max=360, description='Rumbo Inicial:')
out = widgets.interactive_output(williamson_turn, {'rumbo_inicial': ri})
display(widgets.VBox([ri, out]))
```
