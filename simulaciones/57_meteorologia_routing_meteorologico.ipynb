# Enrutamiento Meteorológico Básico
Algoritmo básico para determinar la viabilidad de un rumbo según oleaje y viento.

```
import ipywidgets as widgets
from IPython.display import display

def evaluar_ruta(viento_knts, dir_viento, altura_olas, angulo_ruta):
    # Angulo relativo del viento
    relativo = abs(dir_viento - angulo_ruta)
    if relativo > 180:
        relativo = 360 - relativo
        
    print(f"Analizando Ruta a {angulo_ruta}º")
    print(f"Ángulo aparente del mar/viento: {relativo}º")
    
    if relativo < 35:
        print("⚠️ ALERTA: Ciñendo demasiado o directamente contra el viento. Velocidad muy reducida, pantocazos.")
    elif relativo > 150:
        print("🌬️ Empopada. Posible riesgo de trasluchada accidental.")
    else:
        print("⛵ Rumbo favorable (Través/Largo/Aleta).")
        
    if altura_olas > 2.5 and relativo < 60:
        print("⚠️ ALTO RIESGO: Mar de proa con olas > 2.5m. Peligro estructural y mareo.")
    elif altura_olas > 4.0:
        print("🛑 CONDICIONES EXTREMAS: Buscar refugio.")

vw = widgets.IntSlider(value=15, min=0, max=50, description='Viento (Knt):')
dv = widgets.IntSlider(value=90, min=0, max=360, description='Dir Viento (º):')
ol = widgets.FloatSlider(value=1.5, min=0, max=8, step=0.5, description='Olas (m):')
rt = widgets.IntSlider(value=135, min=0, max=360, description='Ruta (º):')

out = widgets.interactive_output(evaluar_ruta, {'viento_knts': vw, 'dir_viento': dv, 'altura_olas': ol, 'angulo_ruta': rt})
display(widgets.VBox([vw, dv, ol, rt, out]))
```
