# Velocity Made Good (VMG) a Barlovento

El VMG es la velocidad real de acercamiento a un punto exacto en el eje del viento. Ceñir más cerrado baja la velocidad del barco (SOG). Ceñir más abierto (arribar) sube la velocidad, pero recorres más distancia. ¿Cuál es el punto óptimo (Sweet Spot)?


```python
import math
import numpy as np
import matplotlib.pyplot as plt

def calcular_vmg(sog_knots, twa_degrees):
    # VMG = SOG * cos(TWA)
    # TWA: True Wind Angle (Ángulo del Viento Real)
    angulo_rad = math.radians(twa_degrees)
    vmg = sog_knots * math.cos(angulo_rad)
    return vmg

# Datos polares ficticios de un velero ceñiendo
twa_lista = [30, 35, 40, 45, 50, 60]
sog_lista = [3.5, 4.8, 5.7, 6.2, 6.5, 7.0]

vmg_lista = [calcular_vmg(s, t) for s, t in zip(sog_lista, twa_lista)]

plt.figure(figsize=(8,4))
plt.plot(twa_lista, vmg_lista, 'b-o')
plt.title("Curva de VMG en Ceñida")
plt.xlabel("Ángulo Viento Real - TWA (grados)")
plt.ylabel("Velocidad de acercamiento VMG (Nudos)")
plt.grid(True)
plt.show()

# Encontrar el VMG óptimo
best_idx = np.argmax(vmg_lista)
print(f"🏆 Rumbo de ceñida óptimo: {twa_lista[best_idx]}º (VMG: {vmg_lista[best_idx]:.2f} kn)")
```
