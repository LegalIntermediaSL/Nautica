# Cinemática Radar: Ploteo de Blancos
Análisis de Rumbo/Velocidad Relativa y vectores de movimiento.


```python
import numpy as np
import matplotlib.pyplot as plt

# Movimiento Propio
Vr = 15  # nudos
Rr = 45  # grados

# Observaciones del Blanco
t1, marcacion1, dist1 = 0, 90, 8.0
t2, marcacion2, dist2 = 6, 88, 6.0

# Pasamos a polares relativas para cartografiar el ploteo en el radar
def pol2cart(rho, phi):
    x = rho * np.sin(np.radians(phi))
    y = rho * np.cos(np.radians(phi))
    return(x, y)

x1, y1 = pol2cart(dist1, marcacion1)
x2, y2 = pol2cart(dist2, marcacion2)

plt.figure(figsize=(6,6))
plt.plot([0, x1], [0, y1], 'k--', alpha=0.3)
plt.plot([0, x2], [0, y2], 'k--', alpha=0.3)
plt.scatter([x1, x2], [y1, y2], color='red')
plt.annotate('T1', (x1, y1), textcoords="offset points", xytext=(5,5))
plt.annotate('T2', (x2, y2), textcoords="offset points", xytext=(5,5))
plt.plot([x1, x2], [y1, y2], color='red', label="Línea de Movimiento Relativo (LMR)")

plt.xlim(-10, 10)
plt.ylim(-10, 10)
plt.axhline(0, color='black', lw=1)
plt.axvline(0, color='black', lw=1)
plt.title("Ploteo de Pantalla de Radar")
plt.legend()
plt.grid(True)
plt.show()
```
