# 20. Estabilidad Longitudinal y Asiento (Problema del PY)

Este notebook interactivo calcula el cambio en el asiento (trim) y los nuevos calados de proa y popa (Cpp, Cpr) tras trasladar un peso longitudinalmente en el buque. Es un problema clásico del examen de Patrón de Yate (PY).

```python
import numpy as np
import matplotlib.pyplot as plt

def calcular_asiento(L, Cpr_ini, Cpp_ini, Eslora, MCTC, Peso_trasladado, Distancia_traslado, sentido_hacia_proa=True):
    """
    L: Eslora entre perpendiculares (m)
    Cpr_ini: Calado a proa inicial (m)
    Cpp_ini: Calado a popa inicial (m)
    MCTC: Momento para cambiar el asiento 1 cm (ton·m)
    Peso_trasladado: (ton)
    Distancia_traslado: (m)
    sentido_hacia_proa: True si se mueve hacia proa, False hacia popa.
    """
    print("--- CÁLCULO DE ASIENTO TRAS TRASLADO DE PESOS ---")
    print(f"Calados Iniciales: Cpr = {Cpr_ini}m | Cpp = {Cpp_ini}m")
    
    Asiento_inicial = Cpp_ini - Cpr_ini
    print(f"Asiento Inicial (A_ini) = {Asiento_inicial:.3f}m (positivo significa apopado)")
    
    # Momento de Alteración (Momento escorante longitudinal)
    Momento = Peso_trasladado * Distancia_traslado
    print(f"Momento Alterador = {Peso_trasladado}t x {Distancia_traslado}m = {Momento} t·m")
    
    # Alteración del asiento en centimetros y luego en metros
    Alteracion_asiento_cm = Momento / MCTC
    Alteracion_asiento_m = Alteracion_asiento_cm / 100
    
    if sentido_hacia_proa:
        print(f"Alteración del Asiento (a) = -{Alteracion_asiento_m:.3f}m (encabuzamiento)")
        Asiento_final = Asiento_inicial - Alteracion_asiento_m
    else:
        print(f"Alteración del Asiento (a) = +{Alteracion_asiento_m:.3f}m (apopamiento)")
        Asiento_final = Asiento_inicial + Alteracion_asiento_m
    
    print(f"Asiento Final (A_fin) = {Asiento_final:.3f}m")
    
    # Asumiendo centro de flotación en medio de la eslora (L/2)
    # Cambio de calado a proa y popa
    delta_C = Alteracion_asiento_m / 2
    
    if sentido_hacia_proa:
        Cpr_fin = Cpr_ini + delta_C
        Cpp_fin = Cpp_ini - delta_C
    else:
        Cpr_fin = Cpr_ini - delta_C
        Cpp_fin = Cpp_ini + delta_C
        
    print(f"Nuevos Calados: Cpr = {Cpr_fin:.3f}m | Cpp = {Cpp_fin:.3f}m")
    
    # --- GRÁFICA ---
    fig, ax = plt.subplots(figsize=(10, 4))
    
    # Perfil del barco inicial
    ax.plot([0, L], [Cpr_ini, Cpp_ini], 'b--', label='Línea de flotación Inicial')
    # Perfil final
    ax.plot([0, L], [Cpr_fin, Cpp_fin], 'r-', label='Línea de flotación Final')
    
    ax.fill_between([0, L], 0, [Cpr_fin, Cpp_fin], color='cyan', alpha=0.2)
    ax.set_ylim([max(Cpp_ini, Cpr_ini, Cpp_fin, Cpr_fin)+2, 0]) # Y invertido
    ax.set_xlim([0, L])
    
    ax.set_xlabel("Eslora (0 = Proa, L = Popa)")
    ax.set_ylabel("Calado (m)")
    ax.set_title("Efecto del Traslado Longitudinal de Pesos en el Calado")
    ax.legend()
    ax.grid(True)
    plt.show()

# EJEMPLO: Eslora 20m, Calados 1.5m proa, 2.0m popa. MCTC 2.5 t·m/cm. 
# Movemos un dinghy de 0.5 ton 10 metros hacia proa.
calcular_asiento(L=20, Cpr_ini=1.5, Cpp_ini=2.0, Eslora=20, MCTC=2.5, Peso_trasladado=0.5, Distancia_traslado=10.0, sentido_hacia_proa=True)
```
