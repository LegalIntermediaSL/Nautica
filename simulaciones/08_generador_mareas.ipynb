# Generador Sinusoidal de Mareas
Aproximación armónica de la Sonda en un instante $t$.


```python
import numpy as np
import matplotlib.pyplot as plt

# Horas en formato decimal para el cálculo matemático
hora_bm = 8.5   # 08:30
hora_pm = 14.75 # 14:45
sonda_bm = 1.2  # metros
sonda_pm = 4.5  # metros

duracion = hora_pm - hora_bm
amplitud = sonda_pm - sonda_bm

# Array de tiempo
tiempos = np.linspace(hora_bm, hora_pm, 100)
# Sonda = S_BM + Amplitud * sin^2(90 * t_transcurrido / duracion)
# Ángulo de fase (0 a 90 grados)
angulo = np.radians(90 * (tiempos - hora_bm) / duracion)
sondas = sonda_bm + amplitud * (np.sin(angulo)**2)

plt.figure(figsize=(8,4))
plt.plot(tiempos, sondas, color='teal', linewidth=3)
plt.scatter([hora_bm, hora_pm], [sonda_bm, sonda_pm], color='red')
plt.text(hora_bm, sonda_bm+0.2, 'BM')
plt.text(hora_pm, sonda_pm+0.2, 'PM')
plt.xlabel("Hora")
plt.ylabel("Sonda (metros)")
plt.title("Curva Paramétrica de Marea (Vaciante/Creciente)")
plt.grid(True)
plt.show()
```
