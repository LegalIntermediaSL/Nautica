# Termodinámica: Predicción de Niebla (Punto de Rocío)

La niebla se forma cuando la temperatura del aire desciende hasta el Punto de Rocío (Dew Point), provocando la condensación del vapor de agua suspendido. Muy común en el mar al caer la tarde (niebla de radiación) o al pasar aire cálido sobre mar muy frío (advección).


```python
import math

def calcular_punto_rocio(temp_c, humedad_relativa):
    # Fórmula aproximada de Magnus-Tetens
    a = 17.27
    b = 237.7
    alpha = ((a * temp_c) / (b + temp_c)) + math.log(humedad_relativa/100.0)
    dew_point = (b * alpha) / (a - alpha)
    return dew_point

t = 18.0
hr = 85.0
pr = calcular_punto_rocio(t, hr)

print(f"Temperatura actual del aire: {t}ºC")
print(f"Humedad Relativa: {hr}%")
print(f"Punto de Rocío (Condensación/Niebla): {pr:.1f}ºC")

if t - pr <= 2.5:
    print("⚠️ ALERTA: Altísima probabilidad de formación inminente de NIEBLA. Prepare radar y acústicas.")
else:
    print("✅ Sin riesgo inminente de niebla por ahora.")
```
