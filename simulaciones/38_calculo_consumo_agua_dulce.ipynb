# Planificador de Provisioning: Agua y Potabilizadora

En un cruce oceánico, la autonomía de agua dulce es crítica. Este script calcula cuándo debes encender la potabilizadora para no quedarte sin reserva.


```python
def logistica_agua(litros_tanque, tripulantes, dias_travesia, l_per_dia=4, cap_potabilizadora_lh=60):
    consumo_diario_total = tripulantes * l_per_dia
    demanda_total = consumo_diario_total * dias_travesia
    
    print(f"=== Plan Logístico: Travesía de {dias_travesia} días con {tripulantes} pax ===")
    print(f"Tanque inicial: {litros_tanque} L")
    print(f"Consumo estimado total: {demanda_total} L ({consumo_diario_total} L/día)")
    
    if demanda_total <= litros_tanque:
        print("✅ Tienes autonomía suficiente sin potabilizadora.")
    else:
        deficit = demanda_total - litros_tanque
        horas_pota = deficit / cap_potabilizadora_lh
        print(f"❌ Faltan {deficit} L. Deberás operar la potabilizadora.")
        print(f"Necesitas {horas_pota:.1f} horas de potabilizadora en total durante la travesía.")
        
logistica_agua(litros_tanque=300, tripulantes=4, dias_travesia=25, l_per_dia=5, cap_potabilizadora_lh=60)
```
