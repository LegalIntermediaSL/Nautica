# Base de Datos Logística: Provisioning Cruce Atlántico (ARC)

¿Cuánta comida embarcar? Un cálculo erróneo puede provocar desnutrición o hundir el barco en sobrepeso. El cálculo estricto divide macros (Hidratos, Proteínas, Grasas, Conservas).


```python
def provisioning_oceanico(pax, dias):
    # Kilos por persona y día recomendados
    agua_litros = 3 * pax * dias  # Sin potabilizadora, crítico.
    hidratos_secos_kg = 0.5 * pax * dias  # Pasta, Arroz, Harina
    proteina_kg = 0.4 * pax * dias # Carne vacio, conservas atún, huevos
    fruta_verdura_fresca = 0.5 * pax * dias # Perecederos
    
    total_kg = agua_litros + hidratos_secos_kg + proteina_kg + fruta_verdura_fresca
    
    print(f"=== Lista de la Compra Oceánica ({pax} Pax / {dias} Días) ===")
    print(f"💧 Agua (Beber/Cocinar): {agua_litros} Litros")
    print(f"🍝 Hidratos secos: {hidratos_secos_kg} kg")
    print(f"🥩 Proteína: {proteina_kg} kg")
    print(f"🍏 Frescos: {fruta_verdura_fresca} kg")
    print("-" * 40)
    print(f"⚖️ Peso extra a estibar en sentinas: {total_kg} Kilos")

provisioning_oceanico(pax=5, dias=21)
```
