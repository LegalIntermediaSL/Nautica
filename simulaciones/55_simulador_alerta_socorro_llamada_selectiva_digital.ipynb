# Llamada Selectiva Digital (DSC)
Simulador del equipo de comunicaciones VHF DSC para emergencias.

Este simulador está diseñado para fines educativos. **No lo utilices para la navegación real.**

```
import ipywidgets as widgets
from IPython.display import display

def dsc_alert(mmsi, categoria, naturaleza, lat, lon):
    print("=== PANTALLA DSC DEL EQUIPO VHF ===")
    print(f"MMSI Propio: {mmsi}")
    print(f"Categoría: {categoria}")
    
    if categoria == "Distress (Socorro)":
        print(f"Naturaleza: {naturaleza}")
        print(f"Posición: {lat} N, {lon} W")
        print("Frecuencia DSC: Canal 70 (156.525 MHz)")
        print("\nPulse [SEND] durante 3-5 segundos para emitir la alerta.")
        print("A continuación, pase al Canal 16 y proceda con el protocolo MAYDAY:")
        print(f"MAYDAY MAYDAY MAYDAY\nAQUÍ EL BARCO {mmsi}...")
    elif categoria == "Urgency (Urgencia)":
        print("Canal DSC: 70")
        print("Protocolo a seguir por voz: PAN PAN, PAN PAN, PAN PAN en Canal 16.")
    elif categoria == "Safety (Seguridad)":
        print("Canal DSC: 70")
        print("Protocolo a seguir por voz: SECURITE, SECURITE, SECURITE en Canal 16.")

mmsi = widgets.Text(value="224123456", description='MMSI:')
categoria = widgets.Dropdown(options=['Distress (Socorro)', 'Urgency (Urgencia)', 'Safety (Seguridad)'], description='Categoría:')
naturaleza = widgets.Dropdown(options=['Undesignated', 'Fire/Explosion', 'Flooding', 'Collision', 'Grounding', 'Listing/Capsizing', 'Sinking', 'Abandoning Ship', 'Piracy', 'Man Overboard'], description='Naturaleza:')
lat = widgets.Text(value="39 12.3", description='Latitud:')
lon = widgets.Text(value="001 02.4", description='Longitud:')

out = widgets.interactive_output(dsc_alert, {'mmsi': mmsi, 'categoria': categoria, 'naturaleza': naturaleza, 'lat': lat, 'lon': lon})
display(widgets.VBox([mmsi, categoria, naturaleza, lat, lon, out]))
```
