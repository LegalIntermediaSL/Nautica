# Cálculo de Consumo y Autonomía
Herramienta para calcular la autonomía en tiempo y distancia según curvas de consumo.

```
import ipywidgets as widgets
from IPython.display import display

def calcular_autonomia(capacidad_tanque, rpm, eslora_flotacion, consumo_base):
    # Velocidad de casco teórica: v = 1.34 * sqrt(LWL)
    # Asumimos que el motor está diseñado para alcanzar la vel. de casco a unas rpm máximas (ej 3000)
    # El consumo no es lineal, es exponencial con respecto a las RPM
    
    rpm_max = 3000
    ratio_rpm = rpm / rpm_max
    
    # Consumo estimado l/h
    consumo_hora = consumo_base * (ratio_rpm ** 2.5)
    
    # Velocidad estimada
    vel_casco = 1.34 * (eslora_flotacion ** 0.5)
    velocidad = vel_casco * ratio_rpm
    
    if velocidad == 0:
        print("Motor parado.")
        return
        
    # Capacidad útil (restando 20% de reserva táctica obligatoria)
    capacidad_util = capacidad_tanque * 0.8
    
    horas_autonomia = capacidad_util / consumo_hora
    millas_autonomia = horas_autonomia * velocidad
    
    print(f"--- Análisis a {rpm} RPM ---")
    print(f"Velocidad Estimada: {velocidad:.1f} nudos (Velocidad de casco: {vel_casco:.1f} nudos)")
    print(f"Consumo Estimado: {consumo_hora:.1f} litros/hora")
    print(f"Capacidad del Tanque: {capacidad_tanque} L (Útil con 20% reserva: {capacidad_util:.1f} L)")
    print(f"Autonomía en Tiempo: {horas_autonomia:.1f} horas")
    print(f"Autonomía en Distancia: {millas_autonomia:.1f} millas náuticas")
    
    if ratio_rpm > 0.8:
        print("\n⚠️ ALERTA: Operando por encima del 80% RPM (Crucero Máximo). El consumo se dispara, reduciendo la autonomía drásticamente.")

tanque = widgets.FloatSlider(value=200, min=20, max=1000, step=10, description='Tanque (L):')
rpm = widgets.IntSlider(value=2000, min=500, max=3000, step=100, description='RPM:')
lwl = widgets.FloatSlider(value=30, min=15, max=50, step=1, description='Eslora Flot.(ft):')
cons_base = widgets.FloatSlider(value=15, min=5, max=50, step=1, description='Cons. Base Max:')

out = widgets.interactive_output(calcular_autonomia, {'capacidad_tanque': tanque, 'rpm': rpm, 'eslora_flotacion': lwl, 'consumo_base': cons_base})
display(widgets.VBox([tanque, rpm, lwl, cons_base, out]))
```
