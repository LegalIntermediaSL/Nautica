# Arquitectura Naval: Obra Viva y Muerta
Calculadora de fuerzas de arrastre y abatimiento basadas en el perfil de la embarcación.

```
import ipywidgets as widgets
from IPython.display import display

def calcular_resistencias(eslora, manga, francobordo, calado, velocidad, viento):
    # Estimación grosera de áreas (m2)
    area_obra_muerta = eslora * francobordo
    area_obra_viva = eslora * calado
    
    # Fuerzas (aprox). Fuerza de arrastre = 0.5 * rho * v^2 * Cd * Area
    rho_aire = 1.225 # kg/m3
    rho_agua = 1025.0 # kg/m3
    
    # Velocidades en m/s
    v_agua_ms = velocidad * 0.51444
    v_viento_ms = viento * 0.51444
    
    # Coeficientes aerodinámicos (estimados)
    cd_aire = 0.8 # Viento lateral
    cd_agua = 0.05 # Resistencia friccional
    
    fuerza_viento_abatimiento = 0.5 * rho_aire * v_viento_ms**2 * cd_aire * area_obra_muerta
    fuerza_arrastre_agua = 0.5 * rho_agua * v_agua_ms**2 * cd_agua * area_obra_viva
    
    print(f"--- Análisis del Perfil del Buque ---")
    print(f"Obra Muerta (Área lateral sobre el agua): ~{area_obra_muerta:.1f} m²")
    print(f"Obra Viva (Área lateral bajo el agua): ~{area_obra_viva:.1f} m²")
    print(f"Ratio Obra Muerta / Obra Viva: {area_obra_muerta/area_obra_viva:.2f}")
    
    if area_obra_muerta/area_obra_viva > 2.0:
        print("⚠️ El barco tiene un perfil muy alto (mucho windage). Abatirá muchísimo con viento lateral.")
    else:
        print("✅ El barco tiene buen agarre en el agua y perfil bajo.")
        
    print(f"\n--- Fuerzas Dinámicas Estimadas ---")
    print(f"Fuerza Lateral por Abatimiento (Viento {viento} kt): {fuerza_viento_abatimiento:.0f} Newtons")
    print(f"Resistencia Friccional (Avance a {velocidad} kt): {fuerza_arrastre_agua:.0f} Newtons")

eslora = widgets.FloatSlider(value=12, min=5, max=100, description='Eslora (m):')
manga = widgets.FloatSlider(value=4, min=2, max=20, description='Manga (m):')
fb = widgets.FloatSlider(value=1.5, min=0.5, max=10, description='Francobordo(m):')
calado = widgets.FloatSlider(value=2, min=0.5, max=10, description='Calado (m):')
vel = widgets.FloatSlider(value=8, min=0, max=30, description='Vel Barco (kt):')
viento = widgets.FloatSlider(value=20, min=0, max=60, description='Viento Lat(kt):')

out = widgets.interactive_output(calcular_resistencias, {'eslora': eslora, 'manga': manga, 'francobordo': fb, 'calado': calado, 'velocidad': vel, 'viento': viento})
display(widgets.VBox([widgets.HBox([eslora, manga]), widgets.HBox([fb, calado]), widgets.HBox([vel, viento]), out]))
```
