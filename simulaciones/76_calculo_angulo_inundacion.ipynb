# Estabilidad: Ángulo de Inundación
Cálculo trigonométrico del Downflooding Angle.

```
import ipywidgets as widgets
from IPython.display import display
import numpy as np

def calcular_inundacion(manga, francobordo, distancia_crujia, altura_apertura):
    # La apertura es un respiradero o escotilla
    # distancia_crujia: a cuántos metros del eje central está (0 = en crujía, manga/2 = en la borda)
    # altura_apertura: altura desde la línea de flotación
    
    if distancia_crujia > manga / 2:
        print("La apertura no puede estar fuera del barco.")
        return
        
    # Calculamos el ángulo en el que la apertura toca el agua
    # Trigometría básica: tan(theta) = Altura / Distancia al eje de giro
    # El eje de giro (centro de flotación) asumimos que está en crujía.
    
    if distancia_crujia == 0:
        # En crujía, no toca el agua hasta escora de 90 grados
        angulo_inundacion = 90
    else:
        rad = np.arctan2(altura_apertura, distancia_crujia)
        angulo_inundacion = np.degrees(rad)
        
    print(f"--- Análisis de Inundación Progresiva ---")
    print(f"Altura de la apertura sobre flotación: {altura_apertura} m")
    print(f"Distancia a crujía: {distancia_crujia} m")
    print(f"\n🌊 ÁNGULO DE INUNDACIÓN (Downflooding Angle): {angulo_inundacion:.1f}º")
    
    if angulo_inundacion < 30:
        print("⚠️ MUY PELIGROSO: La apertura se sumergirá con escoras normales de navegación a vela.")
    elif angulo_inundacion < 50:
        print("⚠️ RIESGO: Si el barco sufre un knock-down, empezará a entrar agua.")
    else:
        print("✅ SEGURO: El ángulo de inundación es amplio.")

m = widgets.FloatSlider(value=4.0, min=2.0, max=10.0, description='Manga (m):')
fb = widgets.FloatSlider(value=1.5, min=0.5, max=5.0, description='Francobordo:')
dc = widgets.FloatSlider(value=1.5, min=0.0, max=5.0, description='Dist a Crujía (m):')
aa = widgets.FloatSlider(value=1.5, min=0.5, max=5.0, description='Alt Apertura (m):')

out = widgets.interactive_output(calcular_inundacion, {'manga': m, 'francobordo': fb, 'distancia_crujia': dc, 'altura_apertura': aa})
display(widgets.VBox([m, fb, dc, aa, out]))
```
