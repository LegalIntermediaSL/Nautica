# Predicción de Mareas por Componentes Armónicas

En lugar del Anuario y interpolación lineal, el método preciso para predecir alturas de marea utiliza curvas cosenoidales de las fuerzas astronómicas (Luna y Sol).


```python
import numpy as np
import matplotlib.pyplot as plt

def marea_armonica(Z0, componentes):
    horas = np.linspace(0, 24, 100)
    altura = np.full_like(horas, Z0) # Nivel medio del mar
    
    for amp, velocidad, fase in componentes:
        # H(t) = Z0 + Sum( A * cos(v*t - fase) )
        rad_v = np.deg2rad(velocidad)
        rad_f = np.deg2rad(fase)
        altura += amp * np.cos(rad_v * horas - rad_f)
        
    plt.figure(figsize=(10,4))
    plt.plot(horas, altura, 'darkblue', linewidth=2)
    plt.title("Curva de Marea (Modelo Armónico Simplificado)")
    plt.xlabel("Hora del día")
    plt.ylabel("Sonda (m)")
    plt.grid(True)
    plt.show()

# Ejemplo ficticio: Z0 = 2.0m. 
# Componentes (Amplitud m, Velocidad º/hora, Fase º)
# M2 (Lunar Semidiurna), S2 (Solar Semidiurna)
comp = [
    (1.2, 28.98, 45),  # M2
    (0.4, 30.00, 30)   # S2
]
marea_armonica(Z0=2.0, componentes=comp)
```
