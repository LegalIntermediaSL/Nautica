# Análisis Vectorial del Viento Aparente
Simulación en tiempo real de cómo afecta la velocidad del barco al ángulo del viento.

```
import ipywidgets as widgets
from IPython.display import display
import matplotlib.pyplot as plt
import math
import numpy as np

def viento_aparente(v_real, dir_real_relativa, v_barco):
    # Pasar a radianes
    ang_rad = math.radians(dir_real_relativa)
    
    # Vector viento real (viniendo HACIA el barco)
    v_real_x = -v_real * math.sin(ang_rad)
    v_real_y = -v_real * math.cos(ang_rad)
    
    # Vector velocidad del barco (crea viento en contra)
    v_barco_x = 0
    v_barco_y = -v_barco
    
    # Vector Viento Aparente
    v_ap_x = v_real_x + v_barco_x
    v_ap_y = v_real_y + v_barco_y
    
    v_ap = math.hypot(v_ap_x, v_ap_y)
    # Angulo aparente viniendo hacia proa (0)
    ang_ap_rad = math.atan2(-v_ap_x, -v_ap_y)
    ang_ap = math.degrees(ang_ap_rad)
    if ang_ap < 0:
        ang_ap += 360
        
    print(f"Velocidad del barco: {v_barco} nudos")
    print(f"Viento Real (TWS): {v_real} nudos desde {dir_real_relativa}º (TWA)")
    print(f"Viento Aparente (AWS): {v_ap:.1f} nudos desde {ang_ap:.1f}º (AWA)")
    
    if v_barco > 0 and dir_real_relativa > 90:
        print("\nObservación: Al ganar velocidad en vientos portantes, el viento aparente se adelanta hacia proa, requiriendo cazar las velas.")
    
    # Graficar Vectores
    fig, ax = plt.subplots(figsize=(6,6))
    ax.set_xlim(-v_real-v_barco, v_real+v_barco)
    ax.set_ylim(-v_real-v_barco, v_real+v_barco)
    ax.axhline(0, color='gray', linestyle='--')
    ax.axvline(0, color='gray', linestyle='--')
    
    # Barco (Vector)
    ax.quiver(0, 0, 0, v_barco, angles='xy', scale_units='xy', scale=1, color='green', label=f'Barco ({v_barco}kts)')
    # Viento Real
    ax.quiver(0, 0, -v_real_x, -v_real_y, angles='xy', scale_units='xy', scale=1, color='blue', label=f'V. Real ({v_real}kts)')
    # Viento Aparente
    ax.quiver(0, 0, -v_ap_x, -v_ap_y, angles='xy', scale_units='xy', scale=1, color='red', label=f'V. Aparente ({v_ap:.1f}kts)')
    
    ax.legend()
    plt.title('Triángulo del Viento Aparente')
    plt.grid()
    plt.show()

vr = widgets.FloatSlider(value=15, min=1, max=40, step=1, description='V. Real (Kt):')
dr = widgets.FloatSlider(value=120, min=0, max=180, step=1, description='Dir. Real (º):')
vb = widgets.FloatSlider(value=8, min=0, max=30, step=0.5, description='Vel. Barco:')

out = widgets.interactive_output(viento_aparente, {'v_real': vr, 'dir_real_relativa': dr, 'v_barco': vb})
display(widgets.VBox([vr, dr, vb, out]))
```
