# Curva de Estabilidad Estática (GZ)
Cálculo hiperbólico simplificado del brazo de adrizamiento.

```
import ipywidgets as widgets
from IPython.display import display
import matplotlib.pyplot as plt
import numpy as np

def plot_gz_curve(gm, angulo_inundacion, angulo_vuelco):
    angulos = np.linspace(0, 90, 100)
    # Ecuación empírica simplificada para fines educativos: GZ = GM * sin(ang) - (constante * ang^3)
    # para forzar el vuelco en angulo_vuelco
    
    a_rad = np.radians(angulos)
    
    # Función hiperbólica simple para simular la curva GZ
    # GZ = GM * sin(theta) + 0.5 * BM * tan^2(theta)*sin(theta) (Fórmula de Wall)
    # Por simplicidad, usamos un modelo de seno amortiguado
    gz = gm * np.sin(a_rad) * (1 - (angulos / angulo_vuelco)**2)
    
    plt.figure(figsize=(8,5))
    plt.plot(angulos, gz, label='Curva de Estabilidad Estática (GZ)')
    plt.axhline(0, color='black', linewidth=1)
    
    if angulo_inundacion < 90:
        plt.axvline(angulo_inundacion, color='blue', linestyle='--', label='Ángulo Inundación')
        
    plt.fill_between(angulos, gz, 0, where=(gz>0) & (angulos<angulo_inundacion), alpha=0.3, color='green', label='Área bajo la curva')
    
    plt.title('Curva GZ (Brazo de Adrizamiento)')
    plt.xlabel('Ángulo de Escora (grados)')
    plt.ylabel('Brazo GZ (metros)')
    plt.legend()
    plt.grid(True)
    plt.show()

gm_in = widgets.FloatSlider(value=0.8, min=0.1, max=3.0, step=0.1, description='GM (m):')
inundacion = widgets.IntSlider(value=45, min=20, max=80, description='Inundación (º):')
vuelco = widgets.IntSlider(value=70, min=50, max=120, description='Vuelco (º):')

out = widgets.interactive_output(plot_gz_curve, {'gm': gm_in, 'angulo_inundacion': inundacion, 'angulo_vuelco': vuelco})
display(widgets.VBox([gm_in, inundacion, vuelco, out]))
```
