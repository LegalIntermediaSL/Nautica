# Navegación Astronómica: Hora de la Meridiana

Para calcular tu Latitud con un sextante (Tomar la Meridiana), necesitas saber a qué hora UTC exacta va a estar el sol en su cenit máximo (meridiano local). Esto depende de tu Longitud (cada 15º = 1 hora) y de la Ecuación del Tiempo.


```python
def hora_meridiana_utc(longitud_grados_w, ecuacion_tiempo_minutos):
    # Paso meridiano del sol medio en Greenwich es a las 12:00 UTC.
    # Corregido por longitud
    hora_media_utc = 12.0 + (longitud_grados_w / 15.0)
    
    # Aplicar la ecuación del tiempo (diferencia entre Sol Verdadero y Sol Medio)
    hora_verdadera_utc = hora_media_utc - (ecuacion_tiempo_minutos / 60.0)
    
    horas = int(hora_verdadera_utc)
    minutos = int((hora_verdadera_utc - horas) * 60)
    
    print(f"Posición observador: Longitud {longitud_grados_w}º W")
    print(f"El Sol cruzará tu meridiano y alcanzará su cota máxima a las:")
    print(f"⏱️ {horas:02d}:{minutos:02d} UTC (¡Ten el sextante preparado unos minutos antes!)")

# Ej: Navegando en el Atlántico (35ºW) en un día donde la Ec. de Tiempo es -4 minutos
hora_meridiana_utc(longitud_grados_w=35.0, ecuacion_tiempo_minutos=-4.0)
```
