# Equilibrio a Vela: Centro Vélico (CV) y Centro de Deriva (CD)

Para que un velero navegue equilibrado y el timonel no sufra fatiga ("caña dura"), el Centro Vélico (CV) debe estar ligeramente a proa del Centro de Deriva (CD) (debido a efectos hidrodinámicos dinámicos conocidos como avance del CD al escorar).


```python
def balance_velero(pos_cv_proa, pos_cd_proa, eslora_flotacion):
    """
    Medidas en metros desde la proa.
    """
    separacion = pos_cd_proa - pos_cv_proa
    lead_percent = (separacion / eslora_flotacion) * 100
    
    print(f"Posición Centro Vélico (CV): {pos_cv_proa} m desde proa")
    print(f"Posición Centro Deriva (CD): {pos_cd_proa} m desde proa")
    print(f"Lead (Distancia CV delante del CD): {separacion:.2f} m")
    print(f"Lead Porcentual (sobre LWL): {lead_percent:.1f}%")
    
    # Criterio estándar de diseño de yates: Lead entre 15% y 20%
    if 15 <= lead_percent <= 20:
        print("⚖️ Velero perfectamente equilibrado.")
    elif lead_percent < 15:
        print("⚠️ Velero ARDIENTE (Tendencia a orzar fuerte). Reducir vela mayor o cazar foque.")
    else:
        print("⚠️ Velero BLANDO (Tendencia a arribar). Riesgoso. Reducir vela de proa.")

balance_velero(pos_cv_proa=4.5, pos_cd_proa=6.1, eslora_flotacion=9.5)
```
