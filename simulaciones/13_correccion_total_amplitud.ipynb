```python
import numpy as np
import matplotlib.pyplot as plt

# --- DATOS DEL EXAMEN ---
# Momento de la observación: 'ORTO' (Amanecer, Este) u 'OCASO' (Atardecer, Oeste)
momento = 'OCASO'

# Nuestra Latitud de Estima (ej: 36º Norte)
latitud = 36.0 

# Declinación del Sol del Almanaque (ej: 20º Sur)
# + para Norte, - para Sur
declinacion_sol = -20.0 

# El azimut que marca nuestra brújula al mirar al sol en el horizonte (Za)
# Ej: El sol se pone y la brújula marca 255º.
za_brujula = 255.0
# ------------------------

# 1. Cálculo de la Amplitud (A)
lat_rad = np.radians(latitud)
dec_rad = np.radians(declinacion_sol)

# sin(A) = sin(dec) / cos(lat)
sin_amp = np.sin(dec_rad) / np.cos(lat_rad)
amplitud_rad = np.arcsin(sin_amp)
amplitud_deg = np.degrees(amplitud_rad)

# 2. Conversión de Amplitud a Azimut Verdadero (Zv)
# La amplitud se cuenta desde el Este (Orto) o el Oeste (Ocaso)
# hacia el Norte (si declinación N) o hacia el Sur (si declinación S).
if momento == 'ORTO':
    punto_cardinal = 90.0 # Este
else:
    punto_cardinal = 270.0 # Oeste

# Si declinación es Norte, el astro sale/se pone más al Norte (restando grados al E o sumando al W)
# Cuidado con los signos, la lógica naval pura:
if momento == 'ORTO':
    zv_calculado = 90.0 - amplitud_deg # Si amp es positiva (N), Zv < 90. Si es negativa (S), Zv > 90.
elif momento == 'OCASO':
    zv_calculado = 270.0 + amplitud_deg # Si amp es positiva (N), Zv > 270. Si es negativa (S), Zv < 270.

# 3. Cálculo de la Corrección Total (Ct)
ct = zv_calculado - za_brujula

print("=== CÁLCULO DE LA CORRECCIÓN TOTAL ===")
print(f"Fase: {momento} | Latitud: {latitud}º | Declinación: {declinacion_sol}º")
print(f"1. Amplitud del Sol: {abs(amplitud_deg):.2f}º hacia el {'N' if declinacion_sol >=0 else 'S'}")
print(f"2. Azimut Verdadero (Zv): {zv_calculado:.1f}º")
print(f"3. Azimut de Aguja (Za): {za_brujula}º")
print("-" * 30)
print(f"CORRECCIÓN TOTAL (Ct): {ct:.1f}º")
print("======================================\n
```
