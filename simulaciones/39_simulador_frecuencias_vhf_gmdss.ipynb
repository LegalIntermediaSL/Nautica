# Simulador GMDSS (VHF vs Inmarsat)

El Sistema Mundial de Socorro (SMSSM/GMDSS) divide el mar en 4 zonas (A1 a A4). Las radios de los barcos deben adaptarse a cada zona para emitir Distress Alerts.


```python
def elegir_radio_gmdss(distancia_costa_mn, latitud):
    print(f"Analizando situación a {distancia_costa_mn} Mn de la costa (Latitud {latitud}º)")
    
    # Simplificación zonas GMDSS
    if distancia_costa_mn <= 30:
        zona = "A1 (Cobertura VHF Costera)"
        radio = "Llamada DSC Canal 70 (VHF). Voz por Canal 16."
    elif distancia_costa_mn <= 150:
        zona = "A2 (Cobertura Onda Media - MF)"
        radio = "Llamada DSC en 2187.5 kHz (MF)."
    elif abs(latitud) <= 70:
        zona = "A3 (Cobertura Satélite Inmarsat)"
        radio = "Inmarsat-C o HF DSC."
    else:
        zona = "A4 (Regiones Polares sin Inmarsat)"
        radio = "HF DSC (Onda Corta) exclusivamente."
        
    print(f"Zona GMDSS operativa: {zona}")
    print(f"Equipo principal a activar para MAYDAY: {radio}\n")

elegir_radio_gmdss(distancia_costa_mn=15, latitud=40)
elegir_radio_gmdss(distancia_costa_mn=120, latitud=40)
elegir_radio_gmdss(distancia_costa_mn=1500, latitud=35) # Pleno Atlántico
```
