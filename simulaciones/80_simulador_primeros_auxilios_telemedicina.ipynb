```
import ipywidgets as widgets
from IPython.display import display

def telemedicina_triaje(sintoma, consciencia, respiracion, hemorragia):
    print("--- TRIAJE MÉDICO EN ALTA MAR ---")
    
    # Evaluación primaria (ABC - Airway, Breathing, Circulation)
    if consciencia == "Inconsciente":
        if respiracion == "No respira":
            print("🚨 CÓDIGO ROJO EXTREMO: RCP INMEDIATA (Reanimación Cardiopulmonar). 30 compresiones / 2 ventilaciones.")
            print("Llamada MAYDAY (Canal 16 / DSC). Solicitud MEDEVAC Urgente.")
            return
        else:
            print("🚨 CÓDIGO ROJO: Paciente inconsciente pero respira. Posición Lateral de Seguridad (PLS).")
            print("Llamada PAN-PAN MÉDICO o MAYDAY. Consulta radio-médica urgente.")
            return
            
    if hemorragia == "Severa / Arterial (Sangre brillante pulsátil)":
        print("🚨 CÓDIGO ROJO: Taponamiento directo inmediato. Si es extremidad y no para, TORNiquete.")
        print("Llamada MAYDAY. MEDEVAC Urgente por shock hipovolémico inminente.")
        return
        
    # Evaluación secundaria
    if sintoma == "Quemadura extensa":
        print("🟠 CÓDIGO NARANJA: Enfriar con agua salada o dulce 15-20 mins. Cubrir con film plástico o gasa vaselinada.")
        print("Peligro de shock e infección. Llama al Centro Radio-Médico para pauta de analgésicos.")
    elif sintoma == "Posible Fractura":
        print("🟠 CÓDIGO NARANJA: Inmovilizar la zona (férula SAM). No intentar colocar el hueso.")
        print("Vigilar pulso distal. Analgésicos del Botiquín C. Si la fractura es abierta, antibioterapia por radio.")
    elif sintoma == "Mareo crónico (Cinetosis severa)":
        print("🟡 CÓDIGO AMARILLO: Peligro de deshidratación severa. Cinarizina / Prometazina.")
        print("Si lleva 24h sin retener líquidos, administrar suero oral a sorbos pequeños o buscar refugio.")
    else:
        print("🟢 CÓDIGO VERDE: Tratamiento local. Vigilar evolución.")

sint = widgets.Dropdown(options=["Mareo crónico (Cinetosis severa)", "Posible Fractura", "Quemadura extensa", "Corte profundo (Hemostasia)"], description='Síntoma:')
cons = widgets.Dropdown(options=["Consciente y orientado", "Confuso", "Inconsciente"], description='Consciencia:')
resp = widgets.Dropdown(options=["Normal", "Dificultosa", "No respira"], description='Respiración:')
hem = widgets.Dropdown(options=["Ninguna o leve", "Severa / Arterial (Sangre brillante pulsátil)"], description='Hemorragia:')

out = widgets.interactive_output(telemedicina_triaje, {'sintoma': sint, 'consciencia': cons, 'respiracion': resp, 'hemorragia': hem})
display(widgets.VBox([cons, resp, hem, sint, out]))
```
