# Simulador Evaluador: Examen Oficial PER (BOE Real Decreto 875/2014)

Este no es un simple script de preguntas. Es el **Motor de Examen Oficial**.
Ejecuta la celda, responde por pantalla, y el sistema corregirá tu examen siguiendo estrictamente los **criterios eliminatorios del Ministerio de Fomento**:

- Total preguntas: 45
- Máximo Errores Totales: **13**
- Tolerancia RIPA (15 preguntas): **Máximo 5 fallos** (Si fallas 6, estás SUSPENSO automático).
- Tolerancia Balizamiento (5 preguntas): **Máximo 2 fallos** (Si fallas 3, SUSPENSO automático).
- Tolerancia Carta Navegación (4 preguntas): **Máximo 2 fallos** (Si fallas 3, SUSPENSO automático).


```python
import random
import time

# Mini base de datos estructurada por módulos restrictivos del PER
BD_PREGUNTAS = {
    "RIPA": [
        {"q": "Un velero y un mercante se cruzan en mar abierto. ¿Quién cede el paso?", 
         "ops": ["a) El velero", "b) El mercante", "c) El de babor", "d) El que vaya más lento"], "ans": "a"},
        {"q": "Si de noche ve una luz verde y una blanca superior, ¿qué es?", 
         "ops": ["a) Un pesquero", "b) Estribor de un buque de propulsión mecánica", "c) Un velero", "d) Remolcador"], "ans": "b"},
        {"q": "Señal acústica de un velero en niebla:", 
         "ops": ["a) 1 larga", "b) 1 larga y 2 cortas", "c) 2 largas", "d) Campana"], "ans": "b"},
        {"q": "Alcanzador y alcanzado: ¿quién maniobra?", 
         "ops": ["a) El alcanzado", "b) El de babor", "c) El alcanzador siempre maniobra", "d) Depende de la vela"], "ans": "c"},
        {"q": "Buque sin gobierno de noche exhibe:", 
         "ops": ["a) Dos rojas en línea vertical", "b) Roja y verde", "c) Blanca y roja", "d) Tres rojas"], "ans": "a"},
        {"q": "Dos veleros a rumbos opuestos con el viento por la misma banda, amurados a babor:", 
         "ops": ["a) Maniobra el de barlovento", "b) Maniobra el de sotavento", "c) Cae a estribor", "d) Cae a babor"], "ans": "a"},
    ],
    "BALIZAMIENTO": [
        {"q": "Marca de peligro aislado (color):", 
         "ops": ["a) Negro y rojo con 2 bolas negras", "b) Verde y rojo", "c) Amarillo y negro", "d) Blanco y rojo"], "ans": "a"},
        {"q": "Marca lateral de Estribor (Región A):", 
         "ops": ["a) Color rojo", "b) Color verde, castillete", "c) Color verde, cónica", "d) Color rojo, cilíndrica"], "ans": "c"},
        {"q": "Marca de Aguas Navegables:", 
         "ops": ["a) Rayas verticales rojas y blancas, bola roja", "b) Cruz amarilla", "c) Rayas horizontales rojas", "d) Negro entero"], "ans": "a"},
    ],
    "CARTA": [
        {"q": "1 Milla Náutica equivale a:", 
         "ops": ["a) 1 minuto de longitud", "b) 1 minuto de latitud (aprox 1852m)", "c) 1 km", "d) 1 grado de latitud"], "ans": "b"},
        {"q": "El Desvío de la Aguja (Delta) depende de:", 
         "ops": ["a) El rumbo del barco", "b) La latitud", "c) El año", "d) La declinación magnética"], "ans": "a"},
        {"q": "Ct (Corrección Total) es igual a:", 
         "ops": ["a) dm - desvio", "b) dm + desvio", "c) dm * desvio", "d) desvio / dm"], "ans": "b"},
    ],
    "GENERAL": [
        {"q": "Si un tripulante cae al agua, la maniobra clásica inmediata es:", 
         "ops": ["a) Boutakov o Anderson", "b) Ciaboga", "c) Fondeo", "d) Abatimiento"], "ans": "a"},
        {"q": "Un chubasco se asocia meteorológicamente a nubes tipo:", 
         "ops": ["a) Cirros", "b) Cúmulonimbos (Cb)", "c) Estratos", "d) Altoestratos"], "ans": "b"},
        {"q": "¿Qué significa la bandera ALFA del Código Internacional?", 
         "ops": ["a) Hombre al agua", "b) Buzo sumergido. Manténgase alejado y a poca velocidad", "c) Práctico a bordo", "d) Peligro de explosión"], "ans": "b"},
        {"q": "Válvula de fondo (pasacascos), es vital mantenerla:", 
         "ops": ["a) Siempre abierta", "b) Cerrada si no se usa, para evitar inundaciones", "c) Lubricada con grasa de litio", "d) Pintada"], "ans": "b"}
    ]
}

def simulador_examen_per():
    print("=" * 60)
    print("🚢 BIENVENIDO AL TRIBUNAL EXAMINADOR DEL P.E.R 🚢")
    print("Real Decreto 875/2014 - Ministerio de Fomento (Simulación Corta)")
    print("=" * 60)
    
    fallos = {"RIPA": 0, "BALIZAMIENTO": 0, "CARTA": 0, "GENERAL": 0}
    
    # Ensamblar examen aleatorio (reducido para Jupyter, pero con lógica real)
    preguntas_examen = []
    for cat in BD_PREGUNTAS:
        # Cogemos todas las de la BBDD de demostración
        for p in BD_PREGUNTAS[cat]:
            preguntas_examen.append((cat, p))
            
    # random.shuffle(preguntas_examen) # Para el examen real
    
    for i, (cat, p) in enumerate(preguntas_examen):
        print(f"\nPregunta {i+1} [{cat}]:")
        print(p["q"])
        for op in p["ops"]: print(op)
        
        # Simulamos la respuesta (para modo interactivo usar input)
        # ans_user = input("Tu respuesta (a/b/c/d): ").lower()
        
        # MODO AUTO-DEMOSTRACIÓN (Para probar el script sin teclear)
        ans_user = random.choice(["a", "b", "c", "d"]) # Respuesta aleatoria para forzar aprobados y suspensos
        # ans_user = p["ans"] # Descomentar para sacar un 10 perfecto
        
        print(f"Tu respuesta: {ans_user}")
        if ans_user == p["ans"]:
            print("✅ Correcta")
        else:
            print(f"❌ Fallo (Correcta: {p['ans']})")
            fallos[cat] += 1
            
        time.sleep(0.3)
        
    # VEREDICTO OFICIAL
    print("\n" + "=" * 60)
    print("📜 ACTA DE EVALUACIÓN OFICIAL")
    print("=" * 60)
    
    fallos_totales = sum(fallos.values())
    print(f"Fallos Totales: {fallos_totales}")
    print(f"- RIPA (Tolerancia max 5): Has fallado {fallos['RIPA']}")
    print(f"- BALIZAMIENTO (Tolerancia max 2): Has fallado {fallos['BALIZAMIENTO']}")
    print(f"- CARTA NAV (Tolerancia max 2): Has fallado {fallos['CARTA']}")
    
    suspenso = False
    motivo = []
    
    # (Para esta demo corta ajustamos los límites matemáticamente a la proporción)
    # Límite RIPA demo: 2
    if fallos['RIPA'] > 2: suspenso = True; motivo.append("Criterio Eliminatorio RIPA")
    if fallos['BALIZAMIENTO'] > 1: suspenso = True; motivo.append("Criterio Eliminatorio Balizamiento")
    if fallos['CARTA'] > 1: suspenso = True; motivo.append("Criterio Eliminatorio Carta Nav")
    if fallos_totales > 6: suspenso = True; motivo.append("Exceso de fallos totales")
    
    if suspenso:
        print("\n❌ RESULTADO: NO APTO (SUSPENSO)")
        print(f"Motivo(s): {', '.join(motivo)}")
        print("Deberá presentarse a la próxima convocatoria ordinaria.")
    else:
        print("\n✅ RESULTADO: APTO (APROBADO)")
        print("Enhorabuena Capitán. Ya puede tramitar su titulación en la Dirección General de la Marina Mercante.")

# Ejecutar el tribunal
simulador_examen_per()
```
