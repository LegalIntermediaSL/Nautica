# Identificador de Astros (Star Finder)

A veces, debido a la nubosidad, solo es visible una estrella durante el crepúsculo y no podemos identificarla a simple vista mirando las constelaciones.

Tomando su **Altura Verdadera ($a_v$)** con el sextante y su **Azimut Verdadero ($Z_v$)** con la pínula del compás, podemos resolver el triángulo de posición *al revés* para hallar su **Declinación ($Dec$)** y su **Ángulo Horario ($P$)**.

Con esos datos astronómicos puros, el ordenador cruzará la información con el catálogo estelar para decirnos de qué estrella se trata.

```python
import math

def deg2rad(deg): return math.radians(deg)
def rad2deg(rad): return math.degrees(rad)

# Catálogo mínimo de las estrellas náuticas más brillantes (Dec, Nombre)
# En un caso real se usa también el Ángulo Sidéreo (AS) calculado desde Aries, 
# pero para esta simulación usaremos la Declinación para un filtrado simple.
catalogo = [
    (89.26, 'Polaris'),
    (38.78, 'Vega'),
    (19.18, 'Arcturus'),
    (-16.71, 'Sirius'),
    (-52.69, 'Canopus'),
    (-60.37, 'Alpha Centauri (Rigil Kentaurus)'),
    (-8.20, 'Rigel'),
    (7.40, 'Betelgeuse'),
    (-29.62, 'Fomalhaut')
]

print("✅ Funciones trigonométricas y catálogo estelar cargados.")
```

## 1. Entrada de Datos de Observación

```python
# --- INPUTS DEL NAVEGANTE ---

l_e = 35.0  # Latitud de Estima (Positiva = Norte, Negativa = Sur)

a_v = 45.5  # Altura Verdadera (grados decimales medidos con el sextante)
Z_v = 135.0 # Azimut Circular (grados decimales leídos en el compás giroscópico)

print(f"Observando un astro a {a_v}º de altura con rumbo {Z_v}º desde Lat {l_e}º.")
```

## 2. Resolución del Triángulo Esférico Inverso

Fórmula de la Cosenusa para la Declinación:
$\sin(Dec) = \sin(l_e) \cdot \sin(a_v) + \cos(l_e) \cdot \cos(a_v) \cdot \cos(Z_v)$

```python
l_rad = deg2rad(l_e)
a_rad = deg2rad(a_v)
z_rad = deg2rad(Z_v)

# Cálculo de Declinación
sin_dec = (math.sin(l_rad) * math.sin(a_rad)) + (math.cos(l_rad) * math.cos(a_rad) * math.cos(z_rad))
sin_dec = max(-1.0, min(1.0, sin_dec))
dec_rad = math.asin(sin_dec)
dec_calc = rad2deg(dec_rad)

print(f"Declinación Matemática de la Estrella: {dec_calc:+.2f}º")
```

## 3. Identificación en el Catálogo

```python
tolerancia = 1.5 # Grados de error permitidos por la estima y refracción
estrella_encontrada = None

for dec_cat, nombre in catalogo:
    if abs(dec_calc - dec_cat) <= tolerancia:
        estrella_encontrada = nombre
        break

print("=== RESULTADO DE IDENTIFICACIÓN ===")
if estrella_encontrada:
    print(f"⭐ ¡Estrella identificada! Se trata de: {estrella_encontrada}")
else:
    print("❌ No se ha encontrado ninguna estrella principal que coincida.")
    print("Puede ser un planeta (cuyas coordenadas cambian a diario) o una estrella de menor magnitud.")
```
