# Generador de Examen Oficial PY
Motor de preguntas aleatorias de navegación y meteo para Patrón de Yate.

```
import ipywidgets as widgets
from IPython.display import display
import random

preguntas_py = [
    {"q": "¿Qué corrige la Corrección Total (Ct)?", "o": ["El Desvío de la aguja", "La Declinación Magnética", "La suma del desvío y la declinación magnética", "El abatimiento"], "a": 2},
    {"q": "¿Cómo se calcula la latitud mediante la polar?", "o": ["Lat = Altura Verdadera + Correcciones de la polar", "Lat = Declinación del sol", "Lat = Altura Instrumental - Desvío", "Requiere usar el almanaque solar"], "a": 0},
    {"q": "¿Qué representa la isobara en un mapa del tiempo?", "o": ["Línea de igual temperatura", "Línea de igual presión atmosférica", "Línea de igual viento", "Línea de corriente marina"], "a": 1},
    {"q": "En el hemisferio Norte, el viento en una borrasca gira:", "o": ["En el sentido de las agujas del reloj", "En sentido contrario a las agujas del reloj", "Hacia el exterior de forma radial", "No hay viento"], "a": 1}
]

def examen_py():
    print("Módulo Interactivo: Simulación Examen Patrón de Yate (PY)")
    print("-------------------------------------------------------")
    for i, p in enumerate(preguntas_py):
        print(f"\nPregunta {i+1}: {p['q']}")
        for j, opt in enumerate(p['o']):
            print(f"  {chr(65+j)}) {opt}")
        print(f"> Respuesta correcta: {chr(65+p['a'])}")

btn = widgets.Button(description="Generar Test Aleatorio PY")
out = widgets.Output()

def on_click(b):
    with out:
        out.clear_output()
        examen_py()

btn.on_click(on_click)
display(btn, out)
```
