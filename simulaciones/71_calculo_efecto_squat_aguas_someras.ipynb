# Efecto Squat (Asentamiento)
Pérdida de calado dinámico al navegar a alta velocidad en canales o aguas someras.

```
import ipywidgets as widgets
from IPython.display import display

def calcular_squat(eslora, manga, calado_estatico, velocidad_barco, profundidad_agua):
    # Fórmula empírica simplificada del Squat Máximo (I.C. Eryuzlu)
    # Squat = Cb * (V^2 / 100) donde V es velocidad
    # Usaremos una aproximación basada en el factor de bloqueo (Blockage factor)
    
    # Coeficiente de bloque asumido para mercante o yate pesado (0.6)
    Cb = 0.6 
    
    if profundidad_agua <= calado_estatico:
        print("El barco ya ha encallado.")
        return
        
    # Relación profundidad / calado
    ratio = profundidad_agua / calado_estatico
    
    if ratio > 5:
        print(f"Ratio = {ratio:.1f}. Aguas suficientemente profundas. El efecto Squat es prácticamente nulo.")
        return
        
    # Squat en metros (Aproximación empírica general)
    squat = (Cb * velocidad_barco**2) / 100
    
    # Factor amplificador si el agua es muy somera
    if ratio < 2:
        squat *= 2 # En aguas confinadas o canales, se duplica
        
    calado_dinamico = calado_estatico + squat
    ukc = profundidad_agua - calado_dinamico
    
    print("--- Análisis Efecto Squat (Asentamiento) ---")
    print(f"Velocidad: {velocidad_barco} nudos")
    print(f"Profundidad real del agua: {profundidad_agua:.2f} m")
    print(f"Calado en parado: {calado_estatico:.2f} m")
    print(f"\nHundimiento dinámico (Squat estimado): {squat:.2f} m")
    print(f"CALADO DINÁMICO TOTAL: {calado_dinamico:.2f} m")
    
    print(f"\nUKC (Under Keel Clearance / Margen bajo la quilla): {ukc:.2f} m")
    
    if ukc < 0:
        print("💥 ALERTA: ENCANDALLAMIENTO INMINENTE. Reduce la velocidad inmediatamente.")
    elif ukc < 0.5:
        print("⚠️ PRECAUCIÓN: UKC crítico. La succión del fondo afectará severamente al gobierno del buque.")

vel = widgets.FloatSlider(value=10, min=2, max=25, description='Velocidad (kt):')
calado = widgets.FloatSlider(value=2.0, min=0.5, max=15, description='Calado (m):')
prof = widgets.FloatSlider(value=4.0, min=1, max=30, description='Profundidad (m):')

out = widgets.interactive_output(calcular_squat, {'eslora': 100, 'manga': 20, 'calado_estatico': calado, 'velocidad_barco': vel, 'profundidad_agua': prof})
display(widgets.VBox([vel, calado, prof, out]))
```
