# Entrenador de Señales Acústicas RIPA (Visibilidad Reducida)

Este script interactivo simula la regla 35 del RIPA. Si ejecutas la función teórica, dictaminará el patrón Morse para diferentes estados de buques en la niebla.


```python
def senal_acustica_niebla(tipo_buque, situacion):
    print(f"Buque: {tipo_buque} | Estado: {situacion}")
    
    if tipo_buque == "Propulsion Mecanica":
        if situacion == "Con Arrancada":
            print("🔊 Señal: 1 Pitada LARGA (cada 2 min máx)")
        elif situacion == "Sin Arrancada (Parado)":
            print("🔊 Señal: 2 Pitadas LARGAS (separadas 2 seg, cada 2 min máx)")
    elif tipo_buque in ["Velero", "Pesquero", "Restringido", "Remolcador"]:
        print("🔊 Señal: 1 Pitada LARGA + 2 CORTAS (cada 2 min máx)")
    elif tipo_buque == "Remolcado":
        print("🔊 Señal: 1 Pitada LARGA + 3 CORTAS (cada 2 min máx, tras la del remolcador)")
    elif tipo_buque == "Fondeado":
        print("🔔 Señal: Repique rápido de CAMPANA durante 5 seg (cada 1 min máx)")
    else:
        print("Desconocido.")
    print("-" * 50)

senal_acustica_niebla("Propulsion Mecanica", "Con Arrancada")
senal_acustica_niebla("Propulsion Mecanica", "Sin Arrancada (Parado)")
senal_acustica_niebla("Velero", "Navegando")
senal_acustica_niebla("Fondeado", "Fondeado")
```
