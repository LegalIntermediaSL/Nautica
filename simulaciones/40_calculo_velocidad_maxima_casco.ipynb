# Hull Speed: Velocidad Máxima del Casco

Para un barco de desplazamiento, es físicamente imposible superar una velocidad teórica dictada por su propia ola de proa y popa (Número de Froude). Intentar acelerar más sólo hunde la popa y consume combustible desmesuradamente.


```python
import math
import numpy as np
import matplotlib.pyplot as plt

def velocidad_desplazamiento(eslora_flotacion_m):
    # Fórmula Hull Speed (nudos) = 1.34 * sqrt(LWL en pies)
    lwl_pies = eslora_flotacion_m * 3.28084
    hull_speed = 1.34 * math.sqrt(lwl_pies)
    return hull_speed

esloras_m = np.linspace(6, 20, 30)
v_max = [velocidad_desplazamiento(l) for l in esloras_m]

plt.figure(figsize=(8,4))
plt.plot(esloras_m, v_max, 'b-o')
plt.title("Velocidad Máxima Teórica (Hull Speed) vs Eslora de Flotación")
plt.xlabel("Eslora en flotación - LWL (metros)")
plt.ylabel("Velocidad Máxima (Nudos)")
plt.grid(True)
plt.show()

eslora_ejemplo = 12 # Velero de 40 pies
print(f"Un barco con eslora en flotación de {eslora_ejemplo}m (aprox 40 pies)")
print(f"Tendrá una velocidad máxima de casco de {velocidad_desplazamiento(eslora_ejemplo):.2f} nudos.")
```
