# Cálculo de Cargas en Fondeo
Tensión hidrodinámica y aerodinámica sobre la línea de fondeo.

```
import ipywidgets as widgets
from IPython.display import display
import math

def calculo_fondeo(eslora, manga, viento_knots, corriente_knots, profundidad, relacion_cadena):
    # Fuerza del viento aproximada en fondeo: F = 1/2 * rho * v^2 * Area_front
    # Simplificación: Fv (kg) = Viento^2 * Manga^2 / 10
    fv = (viento_knots**2) * (manga**2) / 10
    
    # Fuerza de la corriente aproximada: Fc (kg) = V^2 * (Eslora*Calado_aprox) * 2
    calado_aprox = manga / 3
    fc = (corriente_knots**2) * (eslora * calado_aprox) * 2
    
    fuerza_total = fv + fc
    cadena_necesaria = profundidad * relacion_cadena
    
    radio_borneo = cadena_necesaria + eslora
    
    print(f"Fuerza aerodinámica estimada: {fv:.0f} kg")
    print(f"Fuerza hidrodinámica estimada: {fc:.0f} kg")
    print(f"Tensión total en la cadena/ancla: {fuerza_total:.0f} kg")
    print(f"Longitud de cadena filada: {cadena_necesaria:.1f} metros (ratio {relacion_cadena}:1)")
    print(f"Radio máximo de borneo: {radio_borneo:.1f} metros")

es = widgets.FloatSlider(value=12, min=5, max=25, description='Eslora (m):')
ma = widgets.FloatSlider(value=4, min=2, max=8, description='Manga (m):')
vi = widgets.FloatSlider(value=15, min=0, max=50, description='Viento (Knt):')
co = widgets.FloatSlider(value=1.5, min=0, max=5, description='Corriente (Knt):')
pr = widgets.FloatSlider(value=5, min=2, max=20, description='Profundidad:')
rc = widgets.IntSlider(value=4, min=3, max=7, description='Ratio Filado:')

out = widgets.interactive_output(calculo_fondeo, {'eslora': es, 'manga': ma, 'viento_knots': vi, 'corriente_knots': co, 'profundidad': pr, 'relacion_cadena': rc})
display(widgets.VBox([es, ma, vi, co, pr, rc, out]))
```
