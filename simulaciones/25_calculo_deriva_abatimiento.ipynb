# Vector Resultante: Deriva y Abatimiento combinados

Este script calcula el Rumbo Efectivo (Ref) y la Velocidad Efectiva (Vef) combinando tres factores simultáneos:
1.  **Máquina:** Rumbo Verdadero (Rv) y Velocidad de Máquina (Vb).
2.  **Corriente:** Rumbo de la corriente (Rc) e Intensidad horaria (Ihc) [Deriva].
3.  **Viento:** Abatimiento angular provocado por el viento (Ab).


```python
import math
import matplotlib.pyplot as plt

def pol2cart(rho, phi_deg):
    # Rumbo náutico a ángulo matemático
    phi_math = 90 - phi_deg
    phi_rad = math.radians(phi_math)
    return rho * math.cos(phi_rad), rho * math.sin(phi_rad)

def cart2pol(x, y):
    rho = math.sqrt(x**2 + y**2)
    phi_rad = math.atan2(y, x)
    phi_deg = 90 - math.degrees(phi_rad)
    return rho, phi_deg % 360

def navegacion_compuesta(rv, vb, abatimiento, rc, ihc):
    # 1. Aplicar abatimiento para hallar Rumbo en Superficie (Rs)
    rs = rv + abatimiento # Positivo a Estribor, negativo a Babor
    
    # 2. Vector Barco en Superficie
    bx, by = pol2cart(vb, rs)
    
    # 3. Vector Corriente
    cx, cy = pol2cart(ihc, rc)
    
    # 4. Vector Resultante (Efectivo)
    ex = bx + cx
    ey = by + cy
    
    vef, ref = cart2pol(ex, ey)
    
    print(f"Rumbo Verdadero: {rv}º a {vb} nudos")
    print(f"Rumbo Superficie (tras abatimiento {abatimiento}º): {rs}º")
    print(f"Corriente: Hacia {rc}º a {ihc} nudos")
    print(f"--- RESULTANTE ---")
    print(f"Rumbo Efectivo (Ref): {ref:.1f}º")
    print(f"Velocidad Efectiva (Vef): {vef:.1f} nudos")
    
    # Ploteo
    plt.figure(figsize=(6,6))
    plt.quiver(0, 0, bx, by, color='b', angles='xy', scale_units='xy', scale=1, label='Vect. Superficie')
    plt.quiver(bx, by, cx, cy, color='c', angles='xy', scale_units='xy', scale=1, label='Corriente')
    plt.quiver(0, 0, ex, ey, color='r', angles='xy', scale_units='xy', scale=1, label='Vect. Efectivo')
    plt.xlim(0, max(ex, bx)+2)
    plt.ylim(0, max(ey, by)+2)
    plt.grid()
    plt.legend()
    plt.title("Navegación con Deriva y Abatimiento")
    plt.show()

navegacion_compuesta(rv=45, vb=6, abatimiento=5, rc=90, ihc=2)
```
