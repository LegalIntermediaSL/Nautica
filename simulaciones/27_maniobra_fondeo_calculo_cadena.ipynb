# Calculador Óptimo de Fondeo (Catenaria)

Para fondear con seguridad, no basta con la regla de "3 veces la profundidad". El peso de la cadena crea una curva catenaria que amortigua los tirones y hace que el ancla trabaje horizontalmente.
El largo de cadena (Scope) depende de la sonda máxima en pleamar y de las fuerzas esperadas.


```python
import math

def calculo_longitud_cadena(sonda, marea_max, obra_muerta, viento_nudos):
    profundidad_total = sonda + marea_max + obra_muerta
    
    # Reglas empíricas por seguridad de fondeo
    # Vientos < 15knts -> Scope 3:1 o 4:1
    # Vientos 15-30knts -> Scope 5:1
    # Vientos > 30knts -> Scope 7:1
    
    if viento_nudos < 15:
        scope = 4
    elif viento_nudos < 30:
        scope = 5
    else:
        scope = 7
        
    cadena_a_filar = profundidad_total * scope
    
    print(f"Profundidad en pleamar + altura proa: {profundidad_total} m")
    print(f"Viento esperado: {viento_nudos} nudos -> Multiplicador Scope: {scope}:1")
    print(f"==> Cadena Mínima a Filar: {cadena_a_filar:.1f} metros")
    
    # Calcular radio del Círculo de Borneo
    radio_borneo = math.sqrt(cadena_a_filar**2 - profundidad_total**2)
    print(f"Radio del Círculo de Borneo (Sin contar eslora): {radio_borneo:.1f} metros")

calculo_longitud_cadena(sonda=5, marea_max=2.5, obra_muerta=1.5, viento_nudos=25)
```
