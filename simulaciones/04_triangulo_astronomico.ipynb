# Resolución del Triángulo de Posición
Cálculo de la Altura Estimada ($a_e$) y el Azimut ($Z$) mediante trigonometría esférica.


```python
import numpy as np

# Parámetros (modificables)
latitud = 36.0  # Positiva = N
declinacion = 22.0  # Positiva = N
horario_local = 45.0  # hl (grados)

# Paso a radianes
l_rad = np.radians(latitud)
d_rad = np.radians(declinacion)
hl_rad = np.radians(horario_local)

# Cálculo de altura estimada (sen a = sen l * sen d + cos l * cos d * cos hl)
sin_a = np.sin(l_rad)*np.sin(d_rad) + np.cos(l_rad)*np.cos(d_rad)*np.cos(hl_rad)
a_est_rad = np.arcsin(sin_a)
a_est = np.degrees(a_est_rad)

print(f"Altura Estimada (a_e): {a_est:.3f}°")

# Cálculo del Azimut (cotg Z = (tan d * cos l / sen hl) - (sen l / tan hl))
cotg_Z = (np.tan(d_rad)*np.cos(l_rad)/np.sin(hl_rad)) - (np.sin(l_rad)/np.tan(hl_rad))
Z_rad = np.arctan(1/cotg_Z)
Z_deg = np.degrees(Z_rad)
if Z_deg < 0:
    Z_deg += 180  # Ajuste de cuadrante según reglas náuticas

print(f"Azimut (Z): N {Z_deg:.1f}° (Depende del Cuadrante)")
```
