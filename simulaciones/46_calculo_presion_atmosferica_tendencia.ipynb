# Tendencia Barométrica: Predecir el Temporal

Un barómetro no te dice qué tiempo hace, sino qué tiempo hará. Lo que importa no es la presión absoluta (ej. 1010 mb), sino la derivada (tendencia en las últimas 3 horas). Una caída brutal (ciclogénesis) advierte de temporales severos.


```python
def evaluar_tendencia_barometrica(presion_hace_3_horas, presion_actual):
    caida_mb = presion_hace_3_horas - presion_actual
    
    print(f"Hace 3h: {presion_hace_3_horas} mb | Ahora: {presion_actual} mb")
    print(f"Tendencia neta: {caida_mb:+.1f} mb en 3h")
    
    if caida_mb <= 0:
        print("☀️ Presión estable o subiendo. Buen tiempo o mejoría.")
    elif 0.1 <= caida_mb <= 1.5:
        print("🌤️ Caída lenta. Probable llegada de frente cálido lejano.")
    elif 1.6 <= caida_mb <= 3.0:
        print("🌧️ Caída rápida. Aproximación inminente de frente frío. Prepara rizos.")
    elif caida_mb > 3.0:
        print("⚠️ ALERTA: Caída violenta (caída >1 mb/hora). CICLOGÉNESIS EXPLOSIVA. ¡Temporal Inminente!")
    
evaluar_tendencia_barometrica(1015.0, 1010.5)
```
