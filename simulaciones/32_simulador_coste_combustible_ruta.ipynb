# Simulador Económico de Travesía a Motor

Bajar 1 nudo la velocidad de crucero de un barco a motor reduce drásticamente el consumo (l/h), ya que la resistencia hidrodinámica crece exponencialmente.
Este script optimiza financieramente una ruta.


```python
import numpy as np
import matplotlib.pyplot as plt

def simulador_economico(distancia_mn, precio_litro, rpm_array, consumo_lh_array, velocidad_kn_array):
    costes_totales = []
    tiempos = []
    
    print(f"Ruta de {distancia_mn} Millas Náuticas. Precio gasoil: {precio_litro}€/L\n")
    print("RPM  | Velocidad | Consumo  | Tiempo viaje | Coste Total")
    print("-" * 55)
    
    for rpm, lh, kn in zip(rpm_array, consumo_lh_array, velocidad_kn_array):
        horas_viaje = distancia_mn / kn
        litros_totales = lh * horas_viaje
        coste_total = litros_totales * precio_litro
        
        costes_totales.append(coste_total)
        tiempos.append(horas_viaje)
        
        print(f"{rpm:4d} | {kn:6.1f} kn | {lh:5.1f} l/h | {horas_viaje:7.1f} h   | {coste_total:7.2f} €")
        
    plt.figure(figsize=(10, 5))
    plt.plot(velocidad_kn_array, costes_totales, 'ro-', label='Coste Ruta (€)')
    plt.title("Coste de la travesía según la Velocidad")
    plt.xlabel("Velocidad crucero (nudos)")
    plt.ylabel("Coste Total (€)")
    plt.grid(True)
    plt.show()

# Curva empírica de un Trawler de 40 pies (ejemplo ficticio)
rpms = [1500, 2000, 2500, 3000, 3500]
vels = [6.0,  8.0,  10.0, 13.0, 16.0]
cons = [5.0,  12.0, 25.0, 50.0, 90.0] 

simulador_economico(distancia_mn=150, precio_litro=1.65, rpm_array=rpms, consumo_lh_array=cons, velocidad_kn_array=vels)
```
