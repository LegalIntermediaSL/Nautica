# Vectores de Atraque con Viento
Uso interactivo de esprines y fuerza propulsora para atracar/desatracar con rachas.

Este simulador está diseñado para fines educativos. **No lo utilices para la navegación real.**

```
import ipywidgets as widgets
from IPython.display import display
import numpy as np

def recomendar_maniobra(direccion_viento, velocidad_viento, lado_atraque):
    # direccion_viento: 0=Proa, 90=Estribor, 180=Popa, 270=Babor
    # lado_atraque: 'Babor' o 'Estribor'
    
    print(f"Situación: Amarrados por {lado_atraque}. Viento de {velocidad_viento} nudos desde {direccion_viento}º relativos.")
    
    v_rad = np.radians(direccion_viento)
    fuerza_x = np.sin(v_rad) * velocidad_viento # Positivo empuja a Estribor, Negativo a Babor
    fuerza_y = np.cos(v_rad) * velocidad_viento # Positivo empuja a Popa, Negativo a Proa
    
    # Lógica de atraque/desatraque
    if velocidad_viento < 5:
        print("\n👉 CONSEJO DESATRAQUE: El viento es flojo. Larga todas las amarras, avante/atrás despacio.")
        return
        
    empuja_al_muelle = False
    if lado_atraque == 'Babor' and fuerza_x < 0:
        empuja_al_muelle = True
    elif lado_atraque == 'Estribor' and fuerza_x > 0:
        empuja_al_muelle = True
        
    if empuja_al_muelle:
        print("\n⚠️ EL VIENTO TE APLASTA CONTRA EL MUELLE (Viento de través hacia tierra).")
        print("👉 CONSEJO DESATRAQUE (Técnica del Spring):")
        if fuerza_y > 0: # Viento de proa-través
            print("1. Deja solo el SPRING DE POPA.")
            print("2. Da máquina AVANTE y mete timón hacia el muelle.")
            print("3. La proa se abrirá al viento. Cuando estés libre, atrás y larga el spring.")
        else: # Viento de popa-través
            print("1. Deja solo el SPRING DE PROA.")
            print("2. Da máquina ATRÁS.")
            print("3. La popa se abrirá al viento. Cuando esté libre, avante y larga el spring.")
    else:
        print("\n✅ EL VIENTO TE ABATE FUERA DEL MUELLE.")
        print("👉 CONSEJO DESATRAQUE: Larga todo, el viento te separará solo. Cuidado con abatir sobre barcos a sotavento.")

v_dir = widgets.FloatSlider(value=45, min=0, max=359, description='Viento (º Rel):')
v_vel = widgets.FloatSlider(value=15, min=0, max=40, description='Vel Viento (kt):')
lado = widgets.Dropdown(options=['Babor', 'Estribor'], description='Atraque:')

out = widgets.interactive_output(recomendar_maniobra, {'direccion_viento': v_dir, 'velocidad_viento': v_vel, 'lado_atraque': lado})
display(widgets.VBox([lado, v_dir, v_vel, out]))
```
