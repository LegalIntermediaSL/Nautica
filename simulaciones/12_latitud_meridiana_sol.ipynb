```python
import matplotlib.pyplot as plt
from matplotlib.patches import Arc

# --- DATOS DEL EXAMEN ---
# Altura Verdadera (av) del Sol al pasar por el meridiano. (Ej: 65º)
av_sol = 65.0 
# El observador miraba hacia el SUR para ver el sol. (Dirección del astro)
mirando_al = 'S' # 'N' o 'S'

# Declinación del Sol sacada del Almanaque Náutico (Ej: 15º Norte)
# Positiva si es Norte, Negativa si es Sur.
declinacion_sol = 15.0  

# ------------------------

# 1. Cálculo de la Distancia Cenital (z)
# z es el ángulo entre el cenit (arriba) y el astro. z = 90 - av
z_valor = 90.0 - av_sol
# El signo de z es contrario al lugar hacia donde miramos el astro.
# Si el sol está al Sur, el cenit 'cae' hacia el Norte respecto al sol.
signo_z = 1 if mirando_al == 'S' else -1
z = z_valor * signo_z

# 2. Cálculo de la Latitud (l = z + declinación)
latitud = z + declinacion_sol

print("=== RESOLUCIÓN ANALÍTICA ===")
print(f"Altura Verdadera (av): {av_sol}º mirando al {mirando_al}")
print(f"Distancia Cenital (z) = 90 - {av_sol} = {z_valor}º ({'N' if signo_z > 0 else 'S'})")
print(f"Declinación (δ): {abs(declinacion_sol)}º ({'N' if declinacion_sol > 0 else 'S'})")
print("-" * 30)
hemisferio = 'Norte (N)' if latitud >= 0 else 'Sur (S)'
print(f"LATITUD OBSERVADA (l): {abs(latitud):.2f}º {hemisferio}")
print("============================\n
```
