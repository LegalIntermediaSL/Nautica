# Efecto de la Hélice (Prop Walk) en Maniobra Atrase

La hélice actúa como una rueda lateral apoyada en el agua, tirando de la popa hacia un lado, especialmente al dar marcha atrás (ciabogar). Saber si tu hélice es levógira o dextrógira es vital para saber atracar.


```python
def efecto_prop_walk(giro_helice, marcha):
    print(f"Hélice: {giro_helice} | Marcha: {marcha}")
    
    if giro_helice == "Dextrógira":
        if marcha == "Avance":
            print("Avance: La popa cae levemente a Estribor (Proa a Babor). Efecto débil.")
        elif marcha == "Atrás":
            print("Atrás: La popa CAE FUERTEMENTE A BABOR. ¡Úsalo para ciabogar o atracar por babor!")
            
    elif giro_helice == "Levógira":
        if marcha == "Avance":
            print("Avance: La popa cae levemente a Babor.")
        elif marcha == "Atrás":
            print("Atrás: La popa CAE FUERTEMENTE A ESTRIBOR. ¡Atraca siempre por estribor al muelle!")

efecto_prop_walk("Dextrógira", "Atrás")
efecto_prop_walk("Levógira", "Atrás")
```
