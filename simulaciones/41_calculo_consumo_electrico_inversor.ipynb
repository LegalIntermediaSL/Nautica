# Cálculo de Consumo Eléctrico 220V vía Inversor

Las baterías de servicio operan a 12V (o 24V). Al conectar un electrodoméstico de 220V (como una cafetera o un secador) a través de un inversor, el consumo de amperios (A) extraído del banco de 12V es gigantesco, debido a la relación de voltajes y a la eficiencia (pérdida térmica) del inversor.


```python
def consumo_baterias_12v(vatios_aparato_220v, eficiencia_inversor=0.85):
    # Potencia (W) = Voltaje (V) * Intensidad (A)
    # Por tanto, Amperios en 12V = W / 12V
    # Hay que añadir la pérdida por el inversor.
    
    amperios_teoricos_12v = vatios_aparato_220v / 12.0
    amperios_reales_12v = amperios_teoricos_12v / eficiencia_inversor
    
    amperios_220v = vatios_aparato_220v / 220.0
    
    print(f"🔌 Aparato de {vatios_aparato_220v}W (Consumo enchufe 220V: {amperios_220v:.1f}A)")
    print(f"🔋 Drenaje sobre banco de 12V (Inversor al {eficiencia_inversor*100}%): {amperios_reales_12v:.1f}A")
    
    if amperios_reales_12v > 100:
        print("⚠️ ¡PELIGRO! Drenaje masivo. Requiere cables de enorme sección para no fundirse.")

# Ej: Microondas de 1200W
consumo_baterias_12v(1200)
```
