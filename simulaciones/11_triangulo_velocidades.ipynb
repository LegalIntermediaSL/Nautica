```python
import numpy as np
import matplotlib.pyplot as plt

# --- VARIABLES DE ENTRADA ---
# Modifica estos valores para ver cómo cambia el Rumbo Efectivo

# Datos del barco
Rumbo_Verdadero = 90  # Grados (0-360), ej: 90º es rumbo Este
Velocidad_Barco = 10  # Nudos (Vb)

# Datos de la corriente
Rumbo_Corriente = 180 # Grados (0-360), ej: 180º es corriente hacia el Sur
Intensidad_Corriente = 4 # Nudos (Ihc)

# -----------------------------

# Función auxiliar para convertir grados (Norte=0, giro horario) a radianes cartesianos
def polar_to_cartesian(angle_deg, magnitude):
    # En matemáticas 0 es el eje X (Este) y gira antihorario.
    # En náutica 0 es el eje Y (Norte) y gira horario.
    angle_rad = np.radians(90 - angle_deg)
    x = magnitude * np.cos(angle_rad)
    y = magnitude * np.sin(angle_rad)
    return x, y

def cartesian_to_polar(x, y):
    magnitude = np.sqrt(x**2 + y**2)
    # math.atan2 devuelve el ángulo desde el eje X
    angle_rad = np.arctan2(y, x)
    # Convertir a ángulo náutico (0 Norte)
    angle_deg = (90 - np.degrees(angle_rad)) % 360
    return magnitude, angle_deg

# 1. Calcular Componentes Vectoriales
x_barco, y_barco = polar_to_cartesian(Rumbo_Verdadero, Velocidad_Barco)
x_corriente, y_corriente = polar_to_cartesian(Rumbo_Corriente, Intensidad_Corriente)

# 2. Suma Vectorial (Rumbo Efectivo)
x_efectivo = x_barco + x_corriente
y_efectivo = y_barco + y_corriente

# 3. Obtener magnitud y ángulo resultante
Vef, Ref = cartesian_to_polar(x_efectivo, y_efectivo)

print("=== RESULTADOS DEL CÁLCULO ===")
print(f"Rumbo Verdadero (Rv): {Rumbo_Verdadero}º a {Velocidad_Barco} nudos.")
print(f"Rumbo Corriente (Rc): {Rumbo_Corriente}º a {Intensidad_Corriente} nudos.")
print("-" * 30)
print(f"RUMBO EFECTIVO (Ref): {Ref:.1f}º")
print(f"VELOCIDAD EFECTIVA (Vef): {Vef:.1f} nudos.")
print("==============================\n")
```
