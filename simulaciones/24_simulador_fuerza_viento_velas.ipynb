# Simulador Aerodinámico: Cuándo tomar rizos

La fuerza ejercida por el viento sobre las velas no crece de forma lineal, sino al **cuadrado de la velocidad del viento aparente**.
Fórmula simplificada de empuje aerodinámico:
$$ F (N) pprox 0.5 \cdot ho \cdot V^2 \cdot S \cdot C_l $$
Donde $ho$ es la densidad del aire (1.225 kg/m³), V es la velocidad en m/s, S es la superficie en m², y $C_l$ el coeficiente de sustentación (aprox 1.0 - 1.2 ciñendo).


```python
import numpy as np
import matplotlib.pyplot as plt

def fuerza_velas(sup_velica_m2, viento_nudos_array, coef_sustentacion=1.1):
    # Conversiones
    rho = 1.225 # kg/m3 (aire a nivel del mar)
    v_ms = viento_nudos_array * 0.51444
    
    # Fuerza en Newtons, paso a Kilos dividiendo entre 9.81
    fuerza_N = 0.5 * rho * (v_ms**2) * sup_velica_m2 * coef_sustentacion
    fuerza_kg = fuerza_N / 9.81
    return fuerza_kg

vientos = np.linspace(5, 40, 50)
f_mayor = fuerza_velas(35, vientos) # Velero medio, 35m2 de vela
f_rizo1 = fuerza_velas(25, vientos) # Con primer rizo
f_rizo2 = fuerza_velas(15, vientos) # Con segundo rizo

plt.figure(figsize=(10,6))
plt.plot(vientos, f_mayor, 'r', label='Mayor completa (35 m²)')
plt.plot(vientos, f_rizo1, 'orange', label='1er Rizo (25 m²)')
plt.plot(vientos, f_rizo2, 'g', label='2do Rizo (15 m²)')
plt.axhline(y=300, color='k', linestyle='--', label='Límite confort/escora (Ej: 300kg)')

plt.title("Fuerza Aerodinámica vs Viento Aparente")
plt.xlabel("Viento Aparente (Nudos)")
plt.ylabel("Fuerza de Empuje/Escora (Kg)")
plt.legend()
plt.grid(True)
plt.show()
```
