# Rutas Transoceánicas: Ortodrómica vs Loxodrómica
Visualización cartográfica y ahorro de millas.


```python
import numpy as np
import matplotlib.pyplot as plt

# Puntos de Salida (A) y Llegada (B)
lat_A, lon_A = 36.0, -5.0   # Estrecho de Gibraltar
lat_B, lon_B = 40.7, -74.0  # Nueva York

# Fórmulas de Loxodrómica y Ortodrómica simplificadas para plotear trayectorias
lats = np.linspace(lat_A, lat_B, 100)
lons_loxo = np.linspace(lon_A, lon_B, 100)

# Ortodrómica (Gran Círculo aproximado para plotear)
# Para un arco perfecto, se requiere trigonometría esférica de interpolación.
# Mostramos una parábola visual en proyeccion mercator
amplitud_orto = 5.0 # Grados de desvío hacia los polos (Hemisferio Norte)
curva_orto = np.sin(np.pi * np.linspace(0, 1, 100)) * amplitud_orto
lats_orto = np.linspace(lat_A, lat_B, 100) + curva_orto

plt.figure(figsize=(10,5))
plt.plot(lons_loxo, lats, label="Loxodrómica (Línea recta en Mercator)", color="red", linestyle="--")
plt.plot(lons_loxo, lats_orto, label="Ortodrómica (Gran Círculo)", color="blue", linewidth=2)
plt.scatter([lon_A, lon_B], [lat_A, lat_B], color='black')
plt.text(lon_A+1, lat_A, 'A', fontsize=12)
plt.text(lon_B+1, lat_B, 'B', fontsize=12)
plt.xlabel("Longitud")
plt.ylabel("Latitud")
plt.title("Navegación Loxodrómica vs Ortodrómica")
plt.legend()
plt.grid(True)
plt.show()
```
