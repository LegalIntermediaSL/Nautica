# 21. Simulador Interactivo VHF (Comunicaciones SMCP)

Las comunicaciones de socorro siguen protocolos rígidos internacionales. Ejecuta la celda para practicar una llamada MAYDAY interactiva. (Es necesario ejecutar el entorno en una terminal que admita input, o bien introducir las respuestas en el popup de Jupyter).

```python
import time

def simulador_mayday():
    print("=== SIMULADOR DE LLAMADA DE SOCORRO (MAYDAY) ===")
    print("El barco 'SIRIUS' (MMSI: 224123456) está sufriendo una vía de agua grave en las coordenadas 39º 12'N, 002º 34'E tras colisionar con un tronco flotante.")
    print("Tu misión es lanzar el MAYDAY por voz a través del canal 16 usando la plantilla SMCP.")
    print("\nPulsa el PTT (Enter) para comenzar...")
    input()
    
    print("\n[Paso 1: Llamada Inicial]")
    print("Completa los espacios para construir la llamada inicial de 3x3x3.")
    p1 = input("Señal de Peligro (repetir 3 veces): ")
    if p1.upper().strip() not in ["MAYDAY MAYDAY MAYDAY", "MAYDAY, MAYDAY, MAYDAY"]:
        print("INCORRECTO. La señal es MAYDAY MAYDAY MAYDAY.")
        return
        
    p2 = input("AQUÍ (seguido del nombre del barco 3 veces): ")
    if "SIRIUS" not in p2.upper():
        print("INCORRECTO. Debías decir: AQUÍ SIRIUS, SIRIUS, SIRIUS.")
        return
        
    print("\n[Paso 2: El Mensaje de Socorro (MIPDA)]")
    print("MIPDA: Mayday, Identificación, Posición, Desgracia, Asistencia (y número de personas).")
    
    p3 = input("MMSI del buque: ")
    if p3 != "224123456":
        print("INCORRECTO. El MMSI era 224123456.")
        return
        
    p4 = input("¿Cuál es tu posición? (Cópiala exacta: 39º 12'N, 002º 34'E): ")
    if "39" not in p4:
        print("INCORRECTO.")
        return
        
    p5 = input("Naturaleza del peligro: ")
    
    print("\nTransmisión completada. Procesando...")
    time.sleep(2)
    print("\n[CENTRO DE SALVAMENTO RECIBE:]")
    print(f"MAYDAY MAYDAY MAYDAY. Aquí SIRIUS, SIRIUS, SIRIUS.")
    print(f"MAYDAY. SIRIUS, MMSI 224123456. Posición {p4}.")
    print(f"{p5}. Solicito asistencia inmediata. Cambio.")
    print("\n¡Excelente! Has completado el ejercicio básico de comunicaciones. Recuerda que la calma y seguir el protocolo MIPDA salvarán vidas.")

# Llama a la función si quieres probarlo (quita el comentario al ejecutar localmente)
# simulador_mayday()
print("Ejecuta la celda y quita el comentario a 'simulador_mayday()' para jugar interactivo.")
```
