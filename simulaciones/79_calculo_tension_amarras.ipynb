# Dinámica de Amarras en Temporal
Carga aerodinámica sobre el buque y cálculo de tensión estructural en las amarras.

```
import ipywidgets as widgets
from IPython.display import display

def tension_amarras(eslora, altura_lateral, velocidad_viento, n_cabos):
    # Cálculo de fuerza lateral (aprox)
    area = eslora * altura_lateral
    viento_ms = velocidad_viento * 0.51444
    rho = 1.225
    
    # Fuerza aerodinámica total (N)
    fuerza_n = 0.5 * rho * viento_ms**2 * 1.0 * area
    
    # Pasar a kilogramos fuerza (kgf)
    fuerza_kgf = fuerza_n / 9.81
    
    # Distribuir la tensión si el viento es de través perfecto
    tension_por_cabo = fuerza_kgf / n_cabos if n_cabos > 0 else 0
    
    print(f"--- Análisis Estático de Amarras (Racha de Través) ---")
    print(f"Viento: {velocidad_viento} kt")
    print(f"Área expuesta (Windage): {area} m²")
    print(f"Fuerza Total sobre el casco: {fuerza_kgf:.0f} kg")
    print(f"Tensión estimada por cabo de trabajo: {tension_por_cabo:.0f} kg")
    
    print("\n🛡️ Elección de Amarra (Factor de Seguridad Recomendado: 1/6 a 1/8 de la Carga de Rotura):")
    carga_rotura_minima = tension_por_cabo * 6
    print(f"Deberías usar cabos con una Carga de Rotura Mínima de: {carga_rotura_minima:.0f} kg")
    
    if carga_rotura_minima > 10000:
        print("⚠️ ALERTA HURACÁN: Las cargas son masivas. Refuerza con estachas dobles, evita el roce (chafing) o sal del puerto si es seguro.")
        
e = widgets.FloatSlider(value=12, min=5, max=30, description='Eslora (m):')
a = widgets.FloatSlider(value=2.5, min=1, max=10, description='Alt Lateral(m):')
v = widgets.FloatSlider(value=40, min=10, max=100, description='Racha (kt):')
c = widgets.IntSlider(value=2, min=1, max=6, description='Cabos a tierra:')

out = widgets.interactive_output(tension_amarras, {'eslora': e, 'altura_lateral': a, 'velocidad_viento': v, 'n_cabos': c})
display(widgets.VBox([e, a, v, c, out]))
```
