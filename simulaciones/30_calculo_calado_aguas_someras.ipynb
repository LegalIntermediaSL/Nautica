# Simulador del Efecto Squat (Asentamiento)

Cuando un buque navega rápido en aguas restringidas o poco profundas, el flujo de agua bajo la quilla se acelera. Por el principio de Bernoulli, la presión disminuye y el barco "se hunde" dinámicamente, perdiendo margen bajo la quilla (UKC - Under Keel Clearance).
