# Error Dinámico del Girocompás (Speed/Latitude Error)

Un girocompás es un disco de inercia electromecánico. Busca el norte verdadero, pero sufre un error de origen cinemático que depende de la latitud, la velocidad del barco y el rumbo. Navegar rápido hacia el Norte/Sur en altas latitudes lo desvía significativamente.


```python
import math

def error_girocompas_rumbo(rumbo_verdadero, velocidad_kn, latitud_grados):
    # Fórmula del Error Dinámico (simplificada, en grados)
    # Error = (Velocidad * cos(Rumbo)) / (900 * cos(Latitud)) 
    # El denominador 900 aproxima (Velocidad de rotación tierra * radio)
    
    lat_rad = math.radians(latitud_grados)
    rumbo_rad = math.radians(rumbo_verdadero)
    
    error_grados = (velocidad_kn * math.cos(rumbo_rad)) / (900 * math.cos(lat_rad))
    # Para convertir a la convención estándar (aprox) usamos el factor de nudos.
    
    # Ajuste para fórmula estándar náutica:
    # Corrección = (V * cos(R)) / (5 \pi cos(L)) -> ~ (V * cos R) / (15 cos L) approx paramétrico real.
    corr_real = -(velocidad_kn * math.cos(rumbo_rad)) / (15.04 * math.cos(lat_rad))
    
    return corr_real

error = error_girocompas_rumbo(rumbo_verdadero=0, velocidad_kn=25, latitud_grados=60)
print(f"Un buque rápido (25 kn) navegando rumbo Norte (000º) en latitud 60ºN")
print(f"Sufrirá un desvío giroscópico de: {error:.2f} grados hacia poniente/levante.")
print("Este error lo compensan automáticamente las giroscópicas modernas si están conectadas a la corredera/GPS.")
```
