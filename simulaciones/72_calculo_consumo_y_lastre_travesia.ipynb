# Evolución de Desplazamiento en Travesía
Cambio en la dinámica del barco al consumir grandes cantidades de combustible.

```
import ipywidgets as widgets
from IPython.display import display

def evolucion_travesia(desplazamiento_inicial, gasoil_inicial_litros, consumo_diario):
    # Peso del gasoil (aprox 0.84 kg/L)
    densidad_gasoil = 0.84
    peso_gasoil_inicial_ton = (gasoil_inicial_litros * densidad_gasoil) / 1000
    
    print(f"Desplazamiento a la salida: {desplazamiento_inicial} Toneladas")
    print(f"Peso del gasoil a bordo: {peso_gasoil_inicial_ton:.2f} Toneladas")
    
    if peso_gasoil_inicial_ton >= desplazamiento_inicial:
        print("Error: El gasoil pesa más que todo el barco.")
        return
        
    dias_max = gasoil_inicial_litros / consumo_diario
    
    print(f"Días de autonomía teórica: {dias_max:.1f} días\n")
    print(f"{'Día':<5} | {'Gasoil Restante (L)':<20} | {'Desplazamiento (Ton)':<20} | {'Efecto'}")
    print("-" * 75)
    
    for dia in range(0, int(dias_max) + 1, 5):
        gasoil_restante = max(0, gasoil_inicial_litros - (consumo_diario * dia))
        peso_gasoil_ton = (gasoil_restante * densidad_gasoil) / 1000
        
        # Desplazamiento actual (Desplazamiento en rosca + peso gasoil actual)
        # Asumiendo que desp inicial ya incluye el gasoil inicial
        desp_actual = desplazamiento_inicial - peso_gasoil_inicial_ton + peso_gasoil_ton
        
        # Un barco más ligero ofrece menos resistencia friccional (puede ir ligerísimamente más rápido o consumir menos)
        variacion_peso = ((desplazamiento_inicial - desp_actual) / desplazamiento_inicial) * 100
        
        efecto = f"El barco es un {variacion_peso:.1f}% más ligero"
        
        print(f"{dia:<5} | {gasoil_restante:<20.0f} | {desp_actual:<20.2f} | {efecto}")

desp = widgets.FloatSlider(value=15.0, min=2.0, max=100.0, step=0.5, description='Desplaz (Ton):')
gasoil = widgets.IntSlider(value=2000, min=100, max=10000, step=100, description='Gasoil (L):')
cons = widgets.IntSlider(value=100, min=10, max=1000, step=10, description='Consumo/Día:')

out = widgets.interactive_output(evolucion_travesia, {'desplazamiento_inicial': desp, 'gasoil_inicial_litros': gasoil, 'consumo_diario': cons})
display(widgets.VBox([desp, gasoil, cons, out]))
```
