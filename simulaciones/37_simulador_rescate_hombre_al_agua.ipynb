# Búsqueda y Salvamento (SAR): Expanding Square Search

El patrón de Cuadrados Expansivos se usa para localizar a un Náufrago (MOB) cuya posición aproximada se conoce, pero hay incertidumbre por la deriva.
Este generador traza matemáticamente la espiral cuadrada.


```python
import matplotlib.pyplot as plt

def generar_expanding_square(radio_inicial_nm, num_piernas):
    # El patrón aumenta su longitud cada dos piernas (legs)
    x, y = [0], [0]
    curr_x, curr_y = 0, 0
    direcciones = [(0, 1), (1, 0), (0, -1), (-1, 0)] # N, E, S, W
    
    distancia = radio_inicial_nm
    dir_idx = 0
    
    for i in range(1, num_piernas + 1):
        dx, dy = direcciones[dir_idx]
        curr_x += dx * distancia
        curr_y += dy * distancia
        x.append(curr_x)
        y.append(curr_y)
        
        # Aumentar la distancia de búsqueda cada 2 giros (90 grados)
        if i % 2 == 0:
            distancia += radio_inicial_nm
            
        dir_idx = (dir_idx + 1) % 4
        
    plt.figure(figsize=(6,6))
    plt.plot(x, y, 'r-o')
    plt.plot(0, 0, 'b*', markersize=15, label='Datum (Última pos conocida)')
    plt.title(f"SAR: Expanding Square Search ({num_piernas} legs)")
    plt.xlabel("Millas Náuticas (X)")
    plt.ylabel("Millas Náuticas (Y)")
    plt.grid()
    plt.legend()
    plt.show()

generar_expanding_square(radio_inicial_nm=0.5, num_piernas=12)
```
