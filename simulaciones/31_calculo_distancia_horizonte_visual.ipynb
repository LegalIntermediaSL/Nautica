# Alcance Visual y Geográfico (Faros)

¿A qué distancia podemos ver la costa o un faro? Depende exclusivamente de nuestra elevación sobre el nivel del mar y de la altura del objeto que miramos.
El alcance geográfico (en millas náuticas) se calcula mediante la fórmula matemática aproximada:
$$ D (mn) = 2.08 \cdot (\sqrt{e} + \sqrt{H}) $$
Donde $e$ es la elevación del ojo del observador (en metros) y $H$ es la altura focal del faro (en metros).


```python
import math
import numpy as np
import matplotlib.pyplot as plt

def alcance_geografico(elevacion_observador_m, altura_faro_m):
    # D = 2.073 * (sqrt(e) + sqrt(h))
    coef = 2.08
    horizonte_observador = coef * math.sqrt(elevacion_observador_m)
    horizonte_faro = coef * math.sqrt(altura_faro_m)
    distancia_total = horizonte_observador + horizonte_faro
    
    print(f"Elevación del observador: {elevacion_observador_m} m")
    print(f"Altura de la luz del faro: {altura_faro_m} m")
    print(f"Alcance de tu horizonte: {horizonte_observador:.2f} Mn")
    print(f"Alcance del horizonte del faro: {horizonte_faro:.2f} Mn")
    print(f"--- Alcance Geográfico Máximo: {distancia_total:.2f} Millas Náuticas ---")
    return distancia_total

# Ejemplo: Faro de Finisterre (143 metros) visto desde la cubierta de un velero (2.5 metros)
alcance_geografico(2.5, 143)

# Gráfico para ver cómo varía según la altura del observador
alturas = np.linspace(1, 25, 50)
alcances = [alcance_geografico(a, 143) for a in alturas]

plt.figure(figsize=(8,4))
plt.plot(alturas, alcances, 'darkblue')
plt.title("Alcance de avistamiento del Faro de Finisterre (143m)")
plt.xlabel("Tu altura de observación (m)")
plt.ylabel("Alcance Geográfico Máximo (Mn)")
plt.grid(True)
plt.show()
```
