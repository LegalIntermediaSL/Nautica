# Navegación Ortodrómica (Círculo Máximo)

La ruta más corta entre dos puntos en una esfera no es una línea recta en la carta de Mercator (loxodrómica), sino un arco de círculo máximo (ortodrómica). 
En grandes travesías oceánicas, la diferencia en millas es sustancial.

Este simulador calcula la distancia ortodrómica, el rumbo inicial, y los vértices.


```python
import math

def deg_to_rad(deg):
    return deg * math.pi / 180

def rad_to_deg(rad):
    return rad * 180 / math.pi

def calculo_ortodromico(lat1, lon1, lat2, lon2):
    # Pasar a radianes
    l1 = deg_to_rad(lat1)
    l2 = deg_to_rad(lat2)
    L = deg_to_rad(lon2 - lon1)
    
    # Distancia ortodrómica
    cos_d = math.sin(l1)*math.sin(l2) + math.cos(l1)*math.cos(l2)*math.cos(L)
    dist_rad = math.acos(cos_d)
    dist_nm = rad_to_deg(dist_rad) * 60
    
    # Rumbo inicial (Ri)
    cos_ri = (math.sin(l2) - math.sin(l1)*cos_d) / (math.cos(l1)*math.sin(dist_rad))
    ri_rad = math.acos(cos_ri)
    ri = rad_to_deg(ri_rad)
    if math.sin(L) < 0:
        ri = 360 - ri
        
    print("--- CÁLCULO ORTODRÓMICO ---")
    print(f"Punto A: Lat {lat1}º, Lon {lon1}º")
    print(f"Punto B: Lat {lat2}º, Lon {lon2}º")
    print(f"Distancia más corta: {dist_nm:.1f} millas náuticas")
    print(f"Rumbo inicial cuadrantal: {ri:.1f}º Verdaderos")
    
# Ejemplo: Finisterre a Nueva York
calculo_ortodromico(lat1=42.88, lon1=-9.26, lat2=40.71, lon2=-74.00)
```
