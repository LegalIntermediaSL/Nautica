# Cálculo del Desvío de la Aguja Magnética (Δ)

El desvío (Δ) es el error angular provocado por los hierros y campos magnéticos propios del barco. Varía según el rumbo al que naveguemos.

Para calcularlo de forma empírica, tomamos la **Demora de Aguja (Da)** a una enfilación cuyo valor verdadero **(Dv)** conocemos por la carta náutica.
Fórmula:
$$ \Delta = Dv - Da - Ct $$
Donde Ct es la Corrección Total. O de forma más simple si trabajamos directo con demora magnética (Dm):
$$ \Delta = Dm - Da $$


```python
def calcular_desvio_aguja(dv, da, dm=None, declinacion_magnetica=0):
    """
    Calcula el desvío de la aguja. 
    Se puede aportar Dv y Declinación Magnética, o directamente Dm.
    """
    if dm is None:
        dm = dv - declinacion_magnetica
        
    desvio = dm - da
    
    # Normalizar entre -180 y 180
    desvio = (desvio + 180) % 360 - 180
    
    signo = "E (+)" if desvio > 0 else "W (-)" if desvio < 0 else "Nulo"
    
    print(f"Demora Verdadera (Dv): {dv}º")
    print(f"Demora Magnética (Dm): {dm}º")
    print(f"Demora de Aguja (Da): {da}º")
    print(f"Desvío Calculado (Δ): {abs(desvio):.2f}º {signo}")
    return desvio

# Ejemplo: Enfilación de dos faros. Carta dice Dv=120º, Declinación mag=-2º (W). Al tomar demora con compás marca 125º.
desvio = calcular_desvio_aguja(dv=120, da=125, declinacion_magnetica=-2)
```
