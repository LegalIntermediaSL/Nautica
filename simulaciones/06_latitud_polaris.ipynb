# Latitud por la Estrella Polar
Aplicando correcciones de Almanaque Náutico: $Lat = a_v - 1^\circ + a_0 + a_1 + a_2$


```python
import matplotlib.pyplot as plt

# Parámetros
a_v = 40.5  # Altura Verdadera de la Polar
a0 = 0.8    # Primera corrección (LHA Aries)
a1 = 0.05   # Segunda corrección (Latitud)
a2 = 0.02   # Tercera corrección (Mes)

latitud_observada = a_v - 1.0 + a0 + a1 + a2
print(f"Altura Verdadera: {a_v}°")
print(f"Latitud Observada: {latitud_observada:.3f}° N")

# Visualización de las correcciones
correcciones = [a_v, -1.0, a0, a1, a2]
labels = ['a_v', '-1° (Constante)', 'a0', 'a1', 'a2']

plt.bar(labels, correcciones, color=['blue', 'red', 'green', 'green', 'green'])
plt.axhline(0, color='black')
plt.ylabel("Grados (°)")
plt.title("Impacto de las correcciones en el cálculo de Polaris")
plt.show()
```
