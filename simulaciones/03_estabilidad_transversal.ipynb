# Estabilidad Transversal
Cálculo gráfico del Momento Adrizante ($M_R$) y movimiento del CM al trasladar pesos.


```python
import numpy as np
import matplotlib.pyplot as plt

# Variables (modificables)
desplazamiento = 5000  # kg
GM = 0.5  # m (Altura metacéntrica inicial)
peso_trasladado = 200  # kg
distancia_traslado = 2.0  # m transversal

# Escora resultante (Aproximación para ángulos pequeños)
# tan(theta) = (w * d) / (Delta * GM)
tan_theta = (peso_trasladado * distancia_traslado) / (desplazamiento * GM)
escora_grados = np.degrees(np.arctan(tan_theta))

print(f"Escora resultante: {escora_grados:.2f}°")

# Curva de establidad (GZ)
# Para ángulos pequeños GZ = GM * sin(theta)
# Para grandes, usar Wall-sided formula: GZ = sin(theta) * (GM + 0.5 * BM * tan^2(theta))
BM = 1.2  # Estimación
angulos = np.linspace(0, 40, 100)
angulos_rad = np.radians(angulos)

GZ = np.sin(angulos_rad) * (GM + 0.5 * BM * np.tan(angulos_rad)**2)
Momento_Adrizante = desplazamiento * GZ

plt.figure(figsize=(8,5))
plt.plot(angulos, Momento_Adrizante, label="Momento Adrizante ($M_R$)", color='b')
plt.axhline(0, color='black', linewidth=1)
plt.xlabel("Escora (°)")
plt.ylabel("Momento Adrizante (kg·m)")
plt.title("Curva de Estabilidad Estática Transversal")
plt.grid(True)
plt.legend()
plt.show()
```
