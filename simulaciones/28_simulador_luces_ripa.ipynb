# Simulador de Escenarios de Luces (RIPA)

¿Qué luces ve el Patrón A al mirar hacia el Patrón B?
Sector Verde (Estribor): 0º a 112.5º
Sector Rojo (Babor): 360º a 247.5º
Sector Blanco (Alcance): 112.5º a 247.5º


```python
def que_luces_veo(rumbo_mio, marcacion_al_otro, rumbo_del_otro):
    # La Demora a la que veo al barco
    demora_al_otro = (rumbo_mio + marcacion_al_otro) % 360
    
    # Desde la perspectiva del barco otro, en qué ángulo relativo estoy yo?
    # Demora opuesta
    demora_desde_el_otro = (demora_al_otro + 180) % 360
    angulo_relativo_suyo = (demora_desde_el_otro - rumbo_del_otro) % 360
    
    # Determinar sector
    if 0 <= angulo_relativo_suyo <= 112.5:
        luces = "VERDE (Costado de Estribor) + TOPE(S)"
        situacion = "Nosotros cruzamos por su ESTRIBOR (Tenemos preferencia según RIPA si ambos a motor)."
    elif 247.5 <= angulo_relativo_suyo <= 360:
        luces = "ROJO (Costado de Babor) + TOPE(S)"
        situacion = "Nosotros cruzamos por su BABOR (Él tiene preferencia, debemos maniobrar)."
    else:
        luces = "BLANCO (Luz de Alcance)"
        situacion = "Le estamos alcanzando (Él tiene preferencia)."
        
    print(f"Vemos sus luces: {luces}")
    print(f"Situación RIPA: {situacion}")

# Ej: Navegamos a 000º. Vemos un barco por la amura de estribor (Marcación 045º). Ese barco navega rumbo 270º.
que_luces_veo(rumbo_mio=0, marcacion_al_otro=45, rumbo_del_otro=270)
```
