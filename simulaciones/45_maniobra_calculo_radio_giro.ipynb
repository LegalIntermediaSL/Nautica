# Curva Evolutiva: Avance y Traslado

Cuando metes todo el timón a una banda, el barco no gira en redondo sobre su centro. Derrapa lateralmente describiendo una curva (Curva Evolutiva). El "Avance" (Advance) es lo que avanza rectilíneamente antes de girar 90º, y el "Traslado" (Transfer) es lo que se ha desplazado lateralmente.


```python
import matplotlib.pyplot as plt
import numpy as np

def simular_curva_evolutiva():
    # Modelado geométrico figurado de una curva evolutiva
    t = np.linspace(0, np.pi, 50)
    
    # Derrape lateral asimétrico 
    x = 100 * np.sin(t) - 20*t
    y = 150 * (1 - np.cos(t))
    
    plt.figure(figsize=(7,5))
    plt.plot(x, y, 'b-', label='Trayectoria Centro Gravedad')
    plt.plot([0, 0], [0, 150], 'r--', label='Línea Rumbo Inicial')
    
    # Marcar 90 grados (Avance y Traslado)
    idx_90 = len(t)//2
    avance = y[idx_90]
    traslado = x[idx_90]
    plt.plot(traslado, avance, 'go', markersize=8)
    
    plt.annotate(f"Avance: {avance:.0f}m", xy=(traslado, avance), xytext=(traslado-150, avance+10))
    plt.annotate(f"Traslado: {abs(traslado):.0f}m", xy=(traslado, avance), xytext=(traslado-150, avance-10))
    
    plt.title("Maniobra: Timón a Babor Toda")
    plt.xlabel("Eje X Lateral (metros)")
    plt.ylabel("Eje Y Proa (metros)")
    plt.grid(True)
    plt.legend()
    plt.show()

simular_curva_evolutiva()
```
