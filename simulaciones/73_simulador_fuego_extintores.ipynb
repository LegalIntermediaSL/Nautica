# Uso de Extintores y Tipos de Fuego
Elección del agente extintor (A, B, C, F) según el conato de incendio en el buque.

Este simulador está diseñado para fines educativos. **No lo utilices para la navegación real.**

```
import ipywidgets as widgets
from IPython.display import display
import random

def simulador_incendio(fuego, extintor):
    print(f"🔥 INCENDIO DETECTADO: {fuego}")
    print(f"🧯 Usando agente extintor: {extintor}\n")
    
    resultado = ""
    if fuego == "Madera, papel, tela (Clase A)":
        if extintor == "Agua (Chorro/Pulverizada)":
            resultado = "✅ ÉXITO: El agua enfría el combustible sólido y apaga el fuego."
        elif extintor == "Polvo Seco (ABC)":
            resultado = "✅ ÉXITO: El polvo aísla el oxígeno y sofoca las llamas. Deja residuo."
        elif extintor == "Espuma":
            resultado = "✅ ÉXITO: La espuma sofoca y enfría ligeramente el fuego."
        else: # CO2
            resultado = "⚠️ PELIGRO: El CO2 sofocará momentáneamente, pero al no enfriar, las brasas pueden reavivarse."
            
    elif fuego == "Gasoil, gasolina, aceite (Clase B)":
        if extintor == "Agua (Chorro/Pulverizada)":
            resultado = "💀 DESASTRE: El agua propagará el líquido inflamable expandiendo el incendio."
        elif extintor == "Polvo Seco (ABC)":
            resultado = "✅ ÉXITO: El polvo ahoga rápidamente el fuego químico."
        elif extintor == "Espuma":
            resultado = "✅ ÉXITO (IDEAL): Crea una capa asfixiante sobre el líquido, sellando los vapores."
        else: # CO2
            resultado = "✅ ÉXITO: Desplaza el oxígeno. Cuidado con asfixiarte en espacios cerrados (sentina)."
            
    elif fuego == "Cuadro eléctrico bajo tensión (Clase C)":
        if extintor == "Agua (Chorro/Pulverizada)" or extintor == "Espuma":
            resultado = "⚡ ELECTROCUCIÓN: El agua y la espuma conducen la electricidad. Riesgo mortal para el tripulante."
        elif extintor == "Polvo Seco (ABC)":
            resultado = "✅ ÉXITO: Apaga el fuego eléctrico, pero el residuo del polvo destruirá toda la electrónica."
        else: # CO2
            resultado = "✅ ÉXITO (IDEAL): No deja residuos, salva la electrónica no afectada."
            
    elif fuego == "Sartén con aceite en cocina (Clase F)":
        if extintor == "Agua (Chorro/Pulverizada)":
            resultado = "💥 EXPLOSIÓN TÉRMICA: El agua evaporará violentamente esparciendo aceite en llamas."
        elif extintor == "Manta Ignífuga":
            resultado = "✅ ÉXITO (IDEAL): Cubrir la sartén cortando el oxígeno es el método más seguro."
        else:
            resultado = "⚠️ INADECUADO: Polvo o CO2 a presión pueden salpicar el aceite ardiente fuera de la sartén."

    print(resultado)

tipos_fuego = ["Madera, papel, tela (Clase A)", "Gasoil, gasolina, aceite (Clase B)", "Cuadro eléctrico bajo tensión (Clase C)", "Sartén con aceite en cocina (Clase F)"]
tipos_extintor = ["Agua (Chorro/Pulverizada)", "Espuma", "Polvo Seco (ABC)", "CO2", "Manta Ignífuga"]

fuego_dd = widgets.Dropdown(options=tipos_fuego, description='Fuego:')
extintor_dd = widgets.Dropdown(options=tipos_extintor, description='Extintor:')

out = widgets.interactive_output(simulador_incendio, {'fuego': fuego_dd, 'extintor': extintor_dd})
display(widgets.VBox([fuego_dd, extintor_dd, out]))
```
