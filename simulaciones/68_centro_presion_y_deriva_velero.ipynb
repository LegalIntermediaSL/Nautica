# Centro Vélico y Resistencia Lateral
Balance físico del velero para evitar tendencias ardientes o blandas.

Este simulador está diseñado para fines educativos. **No lo utilices para la navegación real.**

```
import ipywidgets as widgets
from IPython.display import display
import matplotlib.pyplot as plt

def balance_velico(posicion_tripulacion, trimado_mayor, trimado_foque):
    # Simulación abstracta del Centro Vélico (CE) y Centro de Resistencia Lateral (CLR)
    # Valores de -10 a 10 (Proa a Popa)
    
    clr_base = 0 # Centro del barco
    ce_base = 0
    
    # Mover peso a proa hunde la proa y mueve el CLR a proa
    clr = clr_base - posicion_tripulacion * 0.5
    
    # Sobre-cazar mayor mueve el CE a popa. Sobre-cazar foque mueve el CE a proa
    ce = ce_base + trimado_mayor - trimado_foque
    
    distancia = ce - clr
    
    if distancia > 1.5:
        tendencia = "MUY ARDIENTE (Fuerte tendencia a orzar)"
        color = 'red'
    elif distancia > 0.5:
        tendencia = "ARDIENTE (El barco busca el viento)"
        color = 'orange'
    elif distancia >= -0.5:
        tendencia = "NEUTRO (Bien balanceado, timón a la vía)"
        color = 'green'
    else:
        tendencia = "BLANDO (El barco tiende a arribar, cae a sotavento)"
        color = 'blue'
        
    print(f"Estado del Barco: {tendencia}")
    
    # Graficar
    fig, ax = plt.subplots(figsize=(8, 2))
    ax.plot([-10, 10], [0, 0], 'k-', lw=2) # Eslora
    ax.plot(-10, 0, 'k>', markersize=10) # Proa
    
    ax.plot(clr, -0.1, 'bo', markersize=12, label='CLR (Centro Resistencia Lateral)')
    ax.plot(ce, 0.1, 'ro', markersize=12, label='CE (Centro Vélico)')
    
    ax.annotate(tendencia, xy=(0, 0.5), color=color, ha='center', fontweight='bold')
    
    ax.set_xlim(-12, 12)
    ax.set_ylim(-1, 1)
    ax.axes.get_yaxis().set_visible(False)
    ax.legend()
    plt.title("Balance del Velero (Proa a la izquierda)")
    plt.show()

# Controles
peso = widgets.FloatSlider(value=0, min=-5, max=5, description='Peso (Proa-Popa):')
mayor = widgets.FloatSlider(value=0, min=0, max=5, description='Cazar Mayor:')
foque = widgets.FloatSlider(value=0, min=0, max=5, description='Cazar Foque:')

out = widgets.interactive_output(balance_velico, {'posicion_tripulacion': peso, 'trimado_mayor': mayor, 'trimado_foque': foque})
display(widgets.VBox([peso, mayor, foque, out]))
```
