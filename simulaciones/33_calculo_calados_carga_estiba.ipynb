# TPC y Alteración de Calados al embarcar pesos

El TPC (Toneladas por Centímetro de Inmersión) indica cuántas toneladas hacen hundir el barco 1 centímetro de forma paralela.
Si el peso no se carga exactamente en el Centro de Flotación (CF), además del hundimiento paralelo habrá un asiento (Trim), modificando los calados de proa y popa.


```python
def cambio_calados(calado_inicial_pr, calado_inicial_pp, peso_toneladas, tpc):
    """
    Simulación simplificada de hundimiento isoclínico (sin considerar alteración de asiento).
    """
    cm_inmersion = peso_toneladas / tpc
    metros_inmersion = cm_inmersion / 100.0
    
    calado_final_pr = calado_inicial_pr + metros_inmersion
    calado_final_pp = calado_inicial_pp + metros_inmersion
    
    print(f"Cargando {peso_toneladas} Toneladas...")
    print(f"Hundimiento paralelo: {cm_inmersion:.2f} cm")
    print(f"Calado final Proa: {calado_final_pr:.3f} m")
    print(f"Calado final Popa: {calado_final_pp:.3f} m")
    
# Ej: Llenar depósitos de gasoil (4 toneladas) en un yate con TPC=2.5 T/cm
cambio_calados(calado_inicial_pr=1.80, calado_inicial_pp=1.90, peso_toneladas=4.0, tpc=2.5)
```
