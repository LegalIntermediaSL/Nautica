# Esfuerzo en la Jarcia Firme (Fatiga de Obenques)

Al escorar un velero, el par de adrizamiento (GZ) generado por la quilla es contrarrestado por la tensión en los obenques de barlovento.
Calcularemos de forma simplificada la tensión Tórica (Safe Working Load) al aumentar el viento.


```python
import numpy as np
import matplotlib.pyplot as plt

def tension_obenque_barlovento(rm_max_kgm, manga_m, eslora_m):
    # Modelo hiper-simplificado estático.
    # El Momento Adrizante (RM) tira del mástil hacia barlovento.
    # El obenque resiste. Suponemos anclaje a 1/2 de manga.
    brazo_obenque = manga_m / 2.0
    
    # Tensión = Momento / Brazo (Aproximación bruta sin ángulo)
    tension_base_kg = rm_max_kgm / brazo_obenque
    
    # Margen de seguridad estático (típicamente x3 o x4 para cargas dinámicas y pantocazos)
    carga_rotura_requerida = tension_base_kg * 4 
    return tension_base_kg, carga_rotura_requerida

# Escoras progresivas simuladas por momentos crecientes
momentos_rm = np.linspace(500, 5000, 20) # kg*m
tensiones = [tension_obenque_barlovento(rm, manga_m=3.5, eslora_m=10)[0] for rm in momentos_rm]

plt.figure(figsize=(7,4))
plt.plot(momentos_rm, tensiones, 'black')
plt.title("Tensión en el Obenque de Barlovento vs Momento Escorante")
plt.xlabel("Momento Escorante (RM en kg·m)")
plt.ylabel("Tensión Estática (Kg)")
plt.grid()
plt.show()
```
